````python
df_clean = df.copy()

df_clean = df_clean[["Country", "Currency_Code"]]

df_clean["Country"] = (df_clean["Country"].astype(str).str.strip())

df_clean["Currency_Code"] = (df_clean["Currency_Code"].astype(str).str.strip().str.upper())
````

````python
import re

df_clean = df_clean[
    df_clean["Currency_Code"].str.fullmatch(r"[A-Z]{3}", na=False)
].copy()

num_currencies = df_clean["Currency_Code"].nunique()
print("Number of unique currencies:", num_currencies)
````

````python
from collections import defaultdict

country_to_currency = defaultdict(list)

for _, row in df_clean.iterrows():
    country = row["Country"].lower()
    currency = row["Currency_Code"]

    if currency not in country_to_currency[country]:
        country_to_currency[country].append(currency)

country_to_currency = dict(country_to_currency)

print("Mapping dictionary size (number of countries):", len(country_to_currency))
````

````python
import re
import unicodedata

def normalize_country(name: str) -> str:
    name = name.upper().strip()

    name = unicodedata.normalize("NFKD", name)
    name = "".join(c for c in name if not unicodedata.combining(c))

    name = re.sub(r"\(.*?\)", "", name)

    name = name.replace("|", " AND ")

    name = re.sub(r"\bTHE\b", "", name)

    name = re.sub(r"[^A-Z\s]", "", name)

    name = re.sub(r"\s+", " ", name).strip()

    return name

def generate_aliases(country_name: str):
    aliases = set()

    base = normalize_country(country_name)
    aliases.add(base)

    aliases.add(base.replace(" AND ", " "))

    aliases.add(base.replace(" REPUBLIC OF ", " "))
    aliases.add(base.replace(" DEMOCRATIC ", " "))

    for w in ["REPUBLIC", "STATE", "STATES", "ISLANDS", "ISLAND"]:
        aliases.add(base.replace(w, "").strip())

    return {a for a in aliases if len(a) > 2}

MANUAL_ALIASES = {
    "IRAN": ["IRAN"],
    "UNITED ARAB EMIRATES": ["UAE"],
    "UNITED STATES OF AMERICA": ["USA", "US", "UNITED STATES"],
    "UNITED KINGDOM": ["UK", "GREAT BRITAIN"],
    "RUSSIAN FEDERATION": ["RUSSIA"],
    "KOREA REPUBLIC": ["SOUTH KOREA"],
    "KOREA DEMOCRATIC PEOPLES REPUBLIC": ["NORTH KOREA"],
}

country_currency_map = {}

for country, currencies in country_to_currency.items():
    canonical = normalize_country(country)
    aliases = generate_aliases(country)

    if canonical in MANUAL_ALIASES:
        aliases.update(MANUAL_ALIASES[canonical])

    for a in aliases:
        country_currency_map[a] = currencies
````

````python
def lookup_country(country: str):
    return country_currency_map.get(country.strip())

print("IRAN →", lookup_country("IRAN"))
print("UAE →", lookup_country("UAE"))
````

````python
def build_faq_texts(faq_data):
    faq_texts = []
    for item in faq_data:
        if isinstance(item, str):
            text = item.strip()

        elif isinstance(item, dict):
            q = item.get("question") or item.get("q") or item.get("Question") or ""
            a = item.get("answer") or item.get("a") or item.get("Answer") or ""
            q = str(q).strip()
            a = str(a).strip()

            if q and a:
                text = f"Q: {q}\nA: {a}"
            elif q:
                text = f"Q: {q}"
            elif a:
                text = f"A: {a}"
            else:
                text = str(item)

        else:
            text = str(item).strip()

        if text:
            faq_texts.append(text)

    return faq_texts

faq_texts = build_faq_texts(faq_data)

print("FAQ count:", len(faq_texts))
print("Sample FAQ:\n", faq_texts[0][:400])
````

````python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=120,
    separators=["\n\n", "\n", ".", "!", "?", ",", " ", ""]
)

book_chunks = []
for _, row in df_book.iterrows():
    full_text = str(row["text"])
    chunks = splitter.split_text(full_text)

    for j, ch in enumerate(chunks):
        book_chunks.append({
            "text": ch,
            "source": row.get("source", "World Travel Book.pdf"),
            "doc_id": str(row.get("doc_id", "")),
            "chunk_index": int(row.get("chunk_index", 0)),
            "subchunk_index": j
        })

book_chunks_df = pd.DataFrame(book_chunks)

print("Original book docs:", len(df_book))
print("Book chunks after split:", len(book_chunks_df))
print("\nSample chunk:\n", book_chunks_df.iloc[12]["text"][:400])
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Flight Search</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی از ابزارهای مهمی که تراول‌بات با استفاده از آن به کاربران خدمات می‌دهد، ابزار جستجوی پرواز است. شما باید ابزاری تعریف کنید که پارامترهایی مانند مبدا، مقصد و تاریخ پرواز را از کاربر دریافت کرده و با استفاده از Amadeus API به جستجوی پروازهای مورد نظر بپردازد. این ابزار باید حداقل ۳ پرواز را با اطلاعات کاملی شامل شماره پرواز، کد ایرلاین، قیمت دقیق (به همراه واحد پول)، تاریخ و ساعت حرکت، مدت پرواز و تعداد توقف‌ها به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **تبدیل نام شهر به کد IATA:** ابزار Amadeus بر اساس کدهای یکتای فرودگاه (IATA Code) کار می‌کند. از تابع مپینگی که در بخش قبل نوشته‌اید استفاده کنید تا نام شهرها را به کد فرودگاه تبدیل کنید (مثلاً Tehran → IKA، Dubai → DXB).

- **مدیریت فرمت تاریخ:** Amadeus تاریخ را به فرمت YYYY-MM-DD می‌پذیرد. از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، کاربران فرمت تاریخ را به گونه‌های متفاوتی وارد می‌کنند(برای مثال به صورت متن و نه عدد). شما باید مدیریت فرمت تاریخ در ابزار و System Prompt را مد نظر داشته باشید.

- **مدیریت خطا:** در صورت بروز خطا در جستجو (مثل عدم یافتن پرواز یا خطای API)، پیام خطای مناسب و قابل فهم به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید تا LangGraph بتواند آن را شناسایی کند.

````python
def get_iata_code(city_name: str, city_to_iata: dict):
    if not city_name:
        return None, "City name is empty."

    city_key = city_name.strip().lower()

    if city_key not in city_to_iata:
        return None, f"I could not find an airport code for city '{city_name}'."

    iata_list = city_to_iata[city_key]

    if not iata_list:
        return None, f"No airport codes available for city '{city_name}'."

    if iata_list[0] == "NAN":
        return iata_list[1], None if len(iata_list) > 1 else f"No valid airport codes available for city '{city_name}'."
    return iata_list[0], None
````

````python
from datetime import datetime

def normalize_date(date_str: str):
    if not date_str:
        return None, "Departure date is missing."

    date_str = date_str.strip()

    formats = [
        "%Y-%m-%d",
        "%Y/%m/%d",
        "%d-%m-%Y",
        "%d/%m/%Y",
        "%b %d %Y",
        "%B %d %Y"
    ]

    for fmt in formats:
        try:
            parsed = datetime.strptime(date_str, fmt)
            return parsed.strftime("%Y-%m-%d"), None
        except ValueError:
            continue

    return None, (
        "Invalid date format. Please use formats like "
        "YYYY-MM-DD or 'Jan 10 2026'."
    )
````

````python
from amadeus import Client, ResponseError
import os

amadeus = Client(
    client_id=os.getenv("AMADEUS_CLIENT_ID"),
    client_secret=os.getenv("AMADEUS_CLIENT_SECRET")
)
````

````python
from langchain_core.tools import tool

@tool
def flight_search(
    origin_city: str,
    destination_city: str,
    departure_date: str,
    adults: int = 1,
    max_results: int = 3
):
    """Search flights using Amadeus.

    Args:
        origin_city: Origin city name (e.g., Tehran).
        destination_city: Destination city name (e.g., Dubai).
        departure_date: Date in YYYY-MM-DD or common formats like YYYY/MM/DD.
        adults: Number of adult passengers.
        max_results: Max number of flight offers to return (min 3 recommended).

    Returns:
        A dict containing origin/destination IATA, normalized date, and a list of flight offers.
    """
    origin_iata, error = get_iata_code(origin_city, city_to_iata)
    if error:
        return {"error": error}

    destination_iata, error = get_iata_code(destination_city, city_to_iata)
    if error:
        return {"error": error}

    dep_date, error = normalize_date(departure_date)
    if error:
        return {"error": error}

    from amadeus import ResponseError
    try:
        response = amadeus.shopping.flight_offers_search.get(
            originLocationCode=origin_iata,
            destinationLocationCode=destination_iata,
            departureDate=dep_date,
            adults=adults,
            max=max_results
        )
    except ResponseError:
        return {"error": "Flight search service is currently unavailable. Please try again later."}

    if not response.data:
        return {"error": f"No flights found from {origin_city} to {destination_city} on {dep_date}."}

    flights = []
    for offer in response.data[:max_results]:
        itinerary = offer["itineraries"][0]
        segments = itinerary["segments"]

        flights.append({
            "airline": segments[0]["carrierCode"],
            "flight_number": segments[0]["number"],
            "price": f"{offer['price']['total']} {offer['price']['currency']}",
            "departure_time": segments[0]["departure"]["at"],
            "duration": itinerary["duration"],
            "stops": len(segments) - 1
        })

    return {
        "origin": origin_iata,
        "destination": destination_iata,
        "date": dep_date,
        "results": flights
    }
````

````python
from amadeus import ResponseError

def airport_iata_to_city_iata(iata_code: str):
    try:
        resp = amadeus.reference_data.locations.get(keyword=iata_code, subType="AIRPORT")
        if resp.data:
            city_code = resp.data[0].get("address", {}).get("cityCode")
            if city_code:
                return city_code, None
        return iata_code, None
    except ResponseError:
        return iata_code, "Could not convert airport code to city code."
````

````python
from amadeus import ResponseError

def get_hotel_ids_for_city(city_code: str, limit: int = 50):
    try:
        hotels = amadeus.reference_data.locations.hotels.by_city.get(cityCode=city_code)
        ids = [h["hotelId"] for h in hotels.data[:limit] if "hotelId" in h]
        return ids, hotels.data, None
    except ResponseError as e:
        return [], [], f"Hotel list error: {str(e)}"
````

````python
def try_get_offers(hotel_ids, check_in, check_out, adults=1, want=5, batch_size=3, max_batches=10):
    collected = []

    for b in range(max_batches):
        start = b * batch_size
        end = start + batch_size
        batch = hotel_ids[start:end]
        if not batch:
            break

        try:
            offers = amadeus.shopping.hotel_offers_search.get(
                hotelIds=",".join(batch),
                adults=adults,
                checkInDate=check_in,
                checkOutDate=check_out
            )
            if offers.data:
                collected.extend(offers.data)
                if len(collected) >= want:
                    break

        except ResponseError as e:
            continue

    return collected
````

````python
from amadeus import ResponseError

def pick_hotel_city_code(city_name: str):

    iatas = city_to_iata.get(city_name.strip().lower(), [])
    if not iatas:
        return None, f"No IATA codes found for city '{city_name}'."

    preferred = {"DXB", "DWC"}
    ordered = sorted(iatas, key=lambda x: (x not in preferred, x))

    last_error = None
    for code in ordered:
        try:
            test = amadeus.reference_data.locations.hotels.by_city.get(cityCode=code)
            if test.data:
                return code, None
        except ResponseError as e:
            last_error = str(e)
            continue

    return None, f"None of the IATA codes worked for hotel search. Last error: {last_error}"
````

````python
from langchain_core.tools import tool
from datetime import datetime

@tool
def hotel_search(
    destination_city: str,
    check_in: str,
    check_out: str,
    adults: int = 1,
    max_results: int = 5,
    budget_max: float = None
):
    """
    Hotel search using Amadeus (primary) + Tavily fallback.

    Returns hotel name, rating, price per night, and location.
    """
    if not destination_city or not destination_city.strip():
        return {"error": "Destination city is missing."}

    check_in_norm, err = normalize_date(check_in)
    if err:
        return {"error": f"Check-in date error: {err}"}

    check_out_norm, err = normalize_date(check_out)
    if err:
        return {"error": f"Check-out date error: {err}"}

    try:
        nights = (datetime.strptime(check_out_norm, "%Y-%m-%d") - datetime.strptime(check_in_norm, "%Y-%m-%d")).days
        if nights <= 0:
            return {"error": "Check-out date must be after check-in date."}
    except Exception:
        return {"error": "Invalid dates."}

    max_results = max(3, int(max_results))

    city_iata, err = pick_hotel_city_code(destination_city)
    if err:
        return {"error": err}

    hotel_ids, hotels_raw, err = get_hotel_ids_for_city(city_iata, limit=80)
    if err:
        return {"error": err}

    if not hotel_ids:
        return {"error": f"No hotels found for city {destination_city} ({city_iata})."}

    offers_data = try_get_offers(
        hotel_ids=hotel_ids,
        check_in=check_in_norm,
        check_out=check_out_norm,
        adults=adults,
        want=max_results,
        batch_size=3,
        max_batches=12
    )

    if not offers_data:
        try:
            q = f"best hotels in {destination_city} rating price per night address"
            web = tavily_client.search(q, max_results=max_results)
            out = []
            for r in web.get("results", []):
                out.append({
                    "hotel_name": r.get("title"),
                    "rating": None,
                    "price_per_night": "N/A (web source)",
                    "location": destination_city,
                    "source_url": r.get("url"),
                })
            return {
                "destination_city": destination_city,
                "city_code": city_iata,
                "check_in": check_in_norm,
                "check_out": check_out_norm,
                "results": out,
                "note": "No Amadeus hotel offers available; results are from web search."
            }
        except Exception:
            return {"error": "No Amadeus offers found and web fallback failed."}

    results = []
    for item in offers_data:
        hotel = item.get("hotel", {})
        name = hotel.get("name")
        rating = hotel.get("rating")
        address = hotel.get("address", {})
        city = address.get("cityName")
        country = address.get("countryCode")

        offer0 = (item.get("offers") or [{}])[0]
        price = offer0.get("price", {})
        total = price.get("total")
        currency = price.get("currency")

        per_night = None
        if total is not None:
            try:
                per_night = float(total) / nights
            except Exception:
                per_night = None

        if budget_max is not None and per_night is not None and per_night > float(budget_max):
            continue

        results.append({
            "hotel_name": name,
            "rating": rating,
            "price_per_night": (f"{per_night:.2f} {currency}" if per_night is not None else None),
            "location": f"{city or ''} {country or ''}".strip(),
        })

        if len(results) >= max_results:
            break

    if not results:
        return {"error": "No hotels matched your budget/filter criteria."}

    return {
        "destination_city": destination_city,
        "city_code": city_iata,
        "check_in": check_in_norm,
        "check_out": check_out_norm,
        "results": results
    }
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Resturant Search</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی دیگر از ابزارهای مفید تراول‌بات، جستجوی رستوران است. شما باید ابزاری تعریف کنید که نام مقصد را از کاربر دریافت کرده و با استفاده از Tavily Search API به جستجوی بهترین رستوران‌ها و غذاهای محلی در آن مقصد بپردازد. این ابزار باید لیستی از رستوران‌های پیشنهادی را همراه با نام، نوع غذا، توضیحات مختصر و لینک اطلاعات بیشتر به کاربر بازگرداند.

- **استفاده از Tavily Search:** از آنجایی که اطلاعات رستوران‌ها و غذاهای محلی در وب به‌روز و متنوع هستند، از Tavily برای جستجوی این اطلاعات استفاده می‌کنیم.

- **ساخت Query مناسب:** query باید شامل کلمات کلیدی مناسب باشد. مثال: `"best restaurants local food in {destination}"` یا `"top rated restaurants cuisine in {destination}"`.

- **فقط پارامتر Destination:** برخلاف ابزار جستجوی پرواز و هتل، این ابزار فقط نام مقصد را به عنوان ورودی می‌گیرد.

- **مدیریت خطا:** در صورت عدم یافتن نتیجه یا خطای API، پیام واضح و مفیدی به کاربر بازگردانید.

````python
import re

BAD_NAME_PATTERNS = [
    r"\bmap\b", r"\bguide\b", r"\bwhere to eat\b", r"\bbest\b", r"\brestaurants?\b",
    r"\bput these\b", r"\bon your radar\b", r"\bclassic brasseries\b"
]

def clean_restaurant_name(name: str) -> str:
    name = (name or "").strip()

    name = re.sub(r"\s*\d\.\d\s*\(.*?reviews?\)\s*", "", name, flags=re.I)
    name = re.sub(r"\(\s*\d[\d,]*\s*reviews?\s*\)", "", name, flags=re.I)

    name = re.sub(r"\s*\.\s*\d\.\d\s*\.\s*", " ", name)

    name = re.sub(r"^\d+[\.\)]\s*", "", name)

    name = re.sub(r"\s+", " ", name).strip(" -–—:;,. ")

    return name
````

````python
def looks_like_restaurant_name(name: str) -> bool:
    if not name:
        return False
    low = name.lower().strip()

    bad_phrases = [
        "where to eat", "best bistros", "best restaurants", "local eats",
        "top", "guide", "map", "restaurants in", "my favourite"
    ]
    if any(p in low for p in bad_phrases):
        return False

    if len(name) > 45:
        return False

    if low in {"paris", "local eats in paris", "best bistros in paris"}:
        return False

    return True
````

````python
def extract_names_from_text(text: str, max_names: int = 6):
    text = text or ""

    parts = re.split(r"[·•\n;]|(?:\s-\s)|(?:\s\|\s)", text)

    candidates = []
    for p in parts:
        p = p.strip()
        if not p:
            continue

        name = p.split(":")[0].strip()

        name = clean_restaurant_name(name)

        if looks_like_restaurant_name(name):
            candidates.append(name)

    uniq = []
    seen = set()
    for c in candidates:
        key = c.lower()
        if key not in seen:
            seen.add(key)
            uniq.append(c)
        if len(uniq) >= max_names:
            break

    return uniq
````

````python
BAD_DOMAINS = [
    "facebook.com", "reddit.com", "tiktok.com", "instagram.com",
    "youtube.com", "youtu.be"
]
def is_bad_url(url: str) -> bool:
    url = (url or "").lower()
    return any(d in url for d in BAD_DOMAINS)

def build_restaurant_queries(destination: str):
    dest = destination.strip()
    return [
        f"best restaurants in {dest} list of names",
        f"top restaurants {dest} local cuisine restaurant names",
        f"{dest} brasseries bistros recommended restaurants names",
    ]
````

````python
def fetch_page_like_content(url: str):
    try:
        q = f"site:{url}"
        res = tavily_client.search(q, max_results=1)
        results = res.get("results", [])
        if results:
            return (results[0].get("content") or results[0].get("snippet") or "").strip()
    except Exception:
        pass
    return ""
````

````python
import re

def extract_description_from_content(content: str, name: str, max_words: int = 120) -> str:
    content = (content or "").strip()
    if not content:
        return ""

    sentences = re.split(r"(?<=[.!?])\s+", content)
    name_low = name.lower()

    picked = []
    for s in sentences:
        if name_low in s.lower():
            picked.append(s.strip())
        if len(" ".join(picked).split()) >= max_words:
            break

    if not picked:
        return " ".join(content.split()[:80])

    text = " ".join(picked)
    return " ".join(text.split()[:max_words])
````

````python
def get_restaurant_page(destination: str, restaurant_name: str):
    q = f"{restaurant_name} {destination} restaurant"
    res = tavily_client.search(q, max_results=3)

    for r in res.get("results", []):
        url = (r.get("url") or "").lower()
        if any(d in url for d in ["facebook.com", "reddit.com", "tiktok.com", "instagram.com", "youtube.com", "youtu.be"]):
            continue

        snippet = (r.get("content") or r.get("snippet") or "").strip()
        if len(snippet.split()) < 20:
            continue

        desc = " ".join(snippet.split()[:120])
        return (r.get("url") or ""), desc

    return "", ""
````

````python
from langchain_core.tools import tool

@tool
def restaurant_search(destination: str, max_results: int = 5):
    """
    Search for restaurants and local food in a destination using Tavily.

    Input:
      - destination: city name (e.g., Paris)
      - max_results: number of restaurants to return (default 5)

    Output:
      - A dict containing destination and a list of restaurants.
        Each restaurant includes: name, cuisine_type, description, url.
    """
    if not destination or not destination.strip():
        return {"error": "Destination is missing. Please provide a city name like 'Paris'."}

    max_results = max(3, min(int(max_results), 10))
    queries = build_restaurant_queries(destination)

    try:
        raw = []
        for q in queries:
            res = tavily_client.search(q, max_results=5)
            raw.extend(res.get("results", []))
    except Exception:
        return {"error": "Restaurant search service is currently unavailable. Please try again later."}

    raw = [r for r in raw if not is_bad_url(r.get("url", ""))]

    if not raw:
        return {"error": f"No restaurant results found for '{destination}'."}

    out = []
    seen = set()

    from collections import defaultdict
    url_count = defaultdict(int)
    MAX_PER_URL = 1

    for r in raw:
        url = (r.get("url") or "").strip()
        snippet = (r.get("content") or r.get("snippet") or "").strip()

        names = extract_names_from_text(snippet, max_names=6)
        if not names:
            continue

        for name in names:
            if url_count[url] >= MAX_PER_URL:
                continue

            specific_url, specific_desc = get_restaurant_page(destination, name)
            if not specific_url or not specific_desc:
                continue

            key = name.lower()
            if key in seen:
                continue
            seen.add(key)

            low = snippet.lower()
            cuisine_type = "Local / Mixed"
            if "bistro" in low or "brasserie" in low:
                cuisine_type = "French Bistro/Brasserie"
            if "michelin" in low:
                cuisine_type += " (Michelin mentioned)"

            page_content = fetch_page_like_content(url) or snippet
            description = extract_description_from_content(page_content, name, max_words=120)
            if len(description.split()) < 25:
                description = " ".join((snippet or "").split()[:80])

            
            out.append({
                "name": name,
                "cuisine_type": cuisine_type,
                "description": specific_desc,
                "url": specific_url
            })
            url_count[url] += 1

            if len(out) >= max_results:
                break

        if len(out) >= max_results:
            break

    if not out:
        return {"error": "Found web pages but could not reliably extract restaurant names. Try a different destination."}

    return {"destination": destination.strip(), "results": out}
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Get Weather Info</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
ابزار دریافت اطلاعات آب و هوا یکی از ابزارهای حیاتی برای برنامه‌ریزی سفر است. شما باید ابزاری تعریف کنید که نام مقصد و تاریخ مورد نظر را از کاربر دریافت کرده و با استفاده از Tavily Search API اطلاعات پیش‌بینی آب و هوا را برای آن مقصد و تاریخ جستجو کند. این ابزار باید اطلاعاتی مانند دما، وضعیت هوا (آفتابی، بارانی، ابری)، رطوبت، احتمال بارندگی و توصیه‌های لباس‌پوشی را به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **استفاده از Tavily Search:** برای دریافت پیش‌بینی آب و هوای به‌روز از جستجوی وب استفاده می‌کنیم، چون منابع هواشناسی آنلاین اطلاعات دقیق‌تری دارند.

- **ساخت Query مناسب:** query باید شامل نام مقصد و تاریخ باشد. مثال: `"weather forecast {destination} {date}"` یا `"weather in {destination} on {date}"`.

- **پارامترهای ورودی:**
  - `destination`: نام شهر یا مقصد (الزامی)
  - `date`: تاریخ مورد نظر به فرمت های مختلف

- **خروجی کامل:** اطلاعات آب و هوا باید شامل

- **پیش‌بینی چند روزه:** اگر ممکن است، اطلاعات پیش‌بینی چند روز آینده را نیز ارائه دهید.

- **مدیریت خطا:** در صورت عدم دسترسی به اطلاعات آب و هوا یا خطای API، پیام مناسب به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.

**مثال Query:**
```
"weather forecast Dubai December 16 2025"
```

````python
from langchain_core.tools import tool
import re
from datetime import datetime
````

````python
def build_weather_queries(destination: str, date_yyyy_mm_dd: str):
    return [
        f"weather forecast {destination} {date_yyyy_mm_dd}",
        f"weather in {destination} on {date_yyyy_mm_dd} temperature humidity precipitation chance",
        f"{destination} 7 day weather forecast {date_yyyy_mm_dd}",
    ]
````

````python
def parse_weather_text(text: str):
    t = (text or "").strip()
    low = t.lower()

    condition = None
    for key in ["sunny", "clear", "cloudy", "overcast", "rain", "rainy", "showers", "snow", "storm", "windy", "fog"]:
        if key in low:
            condition = key
            break

    temp = None
    temp_unit = None

    m = re.search(r"(-?\d{1,2})\s*(°\s*)?(c|°c|celsius)\b", low)
    if m:
        temp = int(m.group(1))
        temp_unit = "C"
    else:
        m = re.search(r"(-?\d{1,2})\s*(°\s*)?(f|°f|fahrenheit)\b", low)
        if m:
            temp = int(m.group(1))
            temp_unit = "F"

    humidity = None
    mh = re.search(r"humidity\s*(is\s*)?(\d{1,3})\s*%", low)
    if mh:
        humidity = int(mh.group(2))
    else:
        mh = re.search(r"(\d{1,3})\s*%\s*humidity", low)
        if mh:
            humidity = int(mh.group(1))

    precip = None
    mp = re.search(r"(chance of rain|precipitation|rain chance)\s*(is\s*)?(\d{1,3})\s*%", low)
    if mp:
        precip = int(mp.group(3))
    else:
        mp = re.search(r"(\d{1,3})\s*%\s*(chance of rain|rain chance|precipitation)", low)
        if mp:
            precip = int(mp.group(1))

    return {
        "temperature": temp,
        "temperature_unit": temp_unit,
        "condition": condition,
        "humidity": humidity,
        "precipitation_chance": precip,
    }
````

````python
def clothing_advice(temp_c: int | None, condition: str | None, precip: int | None):
    tips = []

    if temp_c is not None:
        if temp_c <= 5:
            tips.append("Wear a heavy jacket/coat, warm layers, and closed shoes.")
        elif temp_c <= 15:
            tips.append("Bring a light jacket and layer-friendly clothing.")
        elif temp_c <= 25:
            tips.append("Comfortable clothing; a light layer for evenings.")
        else:
            tips.append("Light breathable clothes, sunglasses, and stay hydrated.")

    if precip is not None and precip >= 40:
        tips.append("Bring an umbrella or a waterproof jacket.")

    if condition in {"windy", "storm", "fog"}:
        tips.append("Consider windproof outerwear and plan for reduced visibility if needed.")

    if not tips:
        tips.append("Dress in layers and check a reliable forecast source close to your travel time.")

    return tips
````

````python
@tool
def get_weather_info(destination: str, date: str, max_results: int = 3):
    """
    Get weather forecast for a destination and date using Tavily search.

    Inputs:
      - destination: city name (required)
      - date: can be in different formats (e.g., 2026-02-10, Feb 10 2026)

    Output:
      - destination, normalized date, extracted weather info (temp/condition/humidity/precip chance),
        multi-day hints if found, and clothing advice.
    """
    if not destination or not destination.strip():
        return {"error": "Destination is missing. Provide a city like 'Dubai'."}

    date_norm, err = normalize_date(date)
    if err:
        return {"error": f"Date error: {err}"}

    queries = build_weather_queries(destination.strip(), date_norm)

    try:
        merged_text = ""
        sources = []
        for q in queries:
            res = tavily_client.search(q, max_results=max_results)
            for r in res.get("results", []):
                url = r.get("url")
                content = (r.get("content") or r.get("snippet") or "").strip()
                if content:
                    merged_text += "\n" + content
                if url:
                    sources.append(url)
    except Exception:
        return {"error": "Weather service is currently unavailable. Please try again later."}

    if not merged_text.strip():
        return {"error": f"No weather information found for {destination} on {date_norm}."}

    info = parse_weather_text(merged_text)

    temp_c = None
    if info["temperature"] is not None and info["temperature_unit"] == "C":
        temp_c = info["temperature"]
    elif info["temperature"] is not None and info["temperature_unit"] == "F":
        temp_c = round((info["temperature"] - 32) * 5 / 9)

    tips = clothing_advice(temp_c, info["condition"], info["precipitation_chance"])

    multi_day_note = None
    if "7 day" in merged_text.lower() or "7-day" in merged_text.lower() or "weekly forecast" in merged_text.lower():
        multi_day_note = "I also found a multi-day/weekly forecast source; check the links for nearby days."

    return {
        "destination": destination.strip(),
        "date": date_norm,
        "weather": {
            "temperature": info["temperature"],
            "temperature_unit": info["temperature_unit"],
            "condition": info["condition"],
            "humidity": info["humidity"],
            "precipitation_chance": info["precipitation_chance"],
        },
        "clothing_advice": tips,
        "multi_day_note": multi_day_note,
        "sources": list(dict.fromkeys(sources))[:5],
    }
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Get Currency Info</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
ابزار دریافت اطلاعات ارز یکی از ابزارهای ضروری برای مسافرانی است که به کشورهای مختلف سفر می‌کنند. شما باید ابزاری تعریف کنید که با توجه به نام کشور مبدا و مقصد و با استفاده از Tavily Search API نرخ تبدیل فعلی بین این دو ارز را جستجو و به کاربر بازگرداند. توجه داشته باشید از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، شما باید به درستی نام کشورها را به کد ارزی آن‌ها مپ کنید و در نهایت جست و جو را با کد ارزی انجام دهید. میتوانید مدیریت این قسمت را در ابزار و system Prompt انجام دهید.

**نکات مهم پیاده‌سازی:**

- **استفاده از Tavily Search:** برای دریافت نرخ تبدیل ارز به‌روز از جستجوی وب استفاده می‌کنیم تا آخرین نرخ‌های بازار را دریافت کنیم.

- **ساخت Query مناسب:** query باید شامل کدهای ارز باشد. مثال: `"USD to AED exchange rate"` یا `"EUR to JPY conversion rate today"`.

- **پارامترهای ورودی:**
  - نام کشور مبدا و مقصد

- **خروجی کامل:** اطلاعات باید شامل:
  - نرخ تبدیل فعلی (مثلاً: 1 USD = 3.67 AED)


- **پشتیبانی از کدهای ارز متداول:** ابزار باید کدهای استاندارد ISO 4217 را بپذیرد (USD، EUR، GBP، JPY، AED، CNY، INR و ...).

- **مدیریت خطا:**
  - شما باید از تابع mapping که نوشته اید برای map کردن نام هر کشور به کد ارزی مربوطه آن کشور استفاده کنید

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.

**مثال Query:**
```
"United States Dollars to UAE Dirhams exchange rate today"
"Iranian Currency to France Currency conversion rate"
```

**مثال خروجی:**
```
1 USD = 3.67 AED
1 AED = 0.27 USD

```

````python
import re

def is_currency_code(x: str) -> bool:
    if not x:
        return False
    x = x.strip().upper()
    return bool(re.fullmatch(r"[A-Z]{3}", x))
````

````python
def to_currency_code(country_or_code: str):
    if not country_or_code or not country_or_code.strip():
        return None, "Country is missing."

    x = country_or_code.strip()

    if is_currency_code(x):
        return x.upper(), None

    try:
        key = normalize_country(x)
    except Exception:
        key = x.strip().upper()

    codes = None
    if "country_currency_map" in globals():
        codes = country_currency_map.get(key)
    if not codes and "lookup_country" in globals():
        codes = lookup_country(key)

    if not codes:
        return None, f"Could not map '{country_or_code}' to a currency code."

    return codes[0], None 
````

````python
def build_fx_query(from_code: str, to_code: str):
    return [
        f"1 {from_code} to {to_code} exchange rate",
        f"1 {from_code} equals how many {to_code}",
        f"{from_code} {to_code} exchange rate today",
    ]
````

````python
def parse_fx_rate(text: str, from_code: str, to_code: str):
    t = (text or "").replace(",", " ")

    p1 = rf"\b1\s*{from_code}\b\s*(?:=|equals|is)\s*\b([0-9]+(?:\.[0-9]+)?)\b\s*\b{to_code}\b"
    m1 = re.search(p1, t, flags=re.IGNORECASE)
    if m1:
        return float(m1.group(1))

    p2 = rf"\b([0-9]+(?:\.[0-9]+)?)\b\s*\b{to_code}\b\s*(?:per|for)\s*\b1\s*{from_code}\b"
    m2 = re.search(p2, t, flags=re.IGNORECASE)
    if m2:
        return float(m2.group(1))

    return None
````

````python
def is_suspicious_rate(rate: float, from_code: str, to_code: str) -> bool:
    if rate is None or rate <= 0:
        return True
    if from_code != to_code and abs(rate - 1.0) < 1e-12:
        return True
    if rate > 1e6:
        return True

    return False
````

````python
from langchain_core.tools import tool

@tool
def get_currency_info(origin_country: str, destination_country: str, amount: float = 1.0):
    """
    Get live currency conversion rate between two countries/currencies using Tavily search.

    Inputs:
      - origin_country: country name or currency code (e.g., "Iran" or "IRR")
      - destination_country: country name or currency code (e.g., "UAE" or "AED")
      - amount: optional amount to convert (default 1)

    Output:
      - rate + reverse_rate + converted_amount + sources
    """
    sources = []

    from_code, err = to_currency_code(origin_country)
    if err:
        return {"error": err}

    to_code, err = to_currency_code(destination_country)
    if err:
        return {"error": err}

    if from_code == to_code:
        return {"error": "Both currencies are the same."}

    try:
        merged = ""
        for q in build_fx_query(from_code, to_code):
            res = tavily_client.search(q, max_results=3)
            for r in res.get("results", []):
                merged += "\n" + (r.get("content") or r.get("snippet") or "")
                if r.get("url"):
                    sources.append(r["url"])
    except Exception:
        return {"error": "Currency rate service is currently unavailable. Please try again later."}

    rate = parse_fx_rate(merged, from_code, to_code)

    if rate is None or is_suspicious_rate(rate, from_code, to_code):
        try:
            merged_rev = ""
            for q in build_fx_query(to_code, from_code):
                res = tavily_client.search(q, max_results=3)
                for r in res.get("results", []):
                    merged_rev += "\n" + (r.get("content") or r.get("snippet") or "")
                    if r.get("url"):
                        sources.append(r["url"])
        except Exception:
            merged_rev = ""

        reverse_rate = parse_fx_rate(merged_rev, to_code, from_code)
        if reverse_rate is not None and reverse_rate != 0:
            rate = 1 / reverse_rate

    if rate is None or is_suspicious_rate(rate, from_code, to_code):
        return {"error": f"Could not extract a reliable rate for {from_code} to {to_code}. Try again later."}

    reverse = 1 / rate

    return {
        "from_currency": from_code,
        "to_currency": to_code,
        "rate": f"1 {from_code} = {rate:.6f} {to_code}",
        "reverse_rate": f"1 {to_code} = {reverse:.6f} {from_code}",
        "converted_amount": f"{amount} {from_code} = {amount * rate:.6f} {to_code}",
        "sources": list(dict.fromkeys(sources))[:5],
        "note": "Rates are extracted from web sources; small rounding differences may occur."
    }
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">FAQ Search</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
ابزار جستجوی FAQ یکی از ابزارهای کلیدی برای پاسخ به سوالات متداول مسافران است. این ابزار به کاربران اجازه می‌دهد بدون نیاز به جستجوی وب، پاسخ سوالات رایج خود را درباره سفر دریافت کنند. شما باید ابزاری تعریف کنید که از سیستم RAG و LanceDB برای جستجوی معنایی در پایگاه دانش FAQ استفاده کند و مرتبط‌ترین پاسخ‌ها را به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **استفاده از Vector Search:** از LanceDB و embedding های تشکیل شده برای جستجوی معنایی در پایگاه FAQ استفاده کنید. این روش سوالاتی که به شکل‌های مختلف پرسیده می‌شوند را تشخیص می‌دهد.

- **پارامتر ورودی:**
  - `query`: سوال کاربر به زبان طبیعی (الزامی)

- **محدود کردن نتایج:** از `limit(2)` استفاده کنید تا فقط 2 مرتبط‌ترین پرسش و پاسخ برگردانده شود و خروجی خلاصه باشد.

- **فرمت خروجی:** برای هر نتیجه:
  - سوال اصلی از پایگاه (Q:)
  - پاسخ کامل (A:)
  - فرمت واضح و خوانا


- **مدیریت نتایج خالی:** اگر هیچ FAQ مرتبطی یافت نشد، پیام `"No relevant FAQ found."` را برگردانید.

- **مدیریت خطا:** در صورت بروز خطا در جستجو، پیام خطا را به صورت واضح نمایش دهید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید تا LangGraph بتواند آن را شناسایی کند.

**مثال سوال کاربر:**
```
"What documents do I need for international travel?"
"Can I take food on the plane?"
"How early should I arrive at airport?"
```

**مثال خروجی:**
```
Based on our FAQ:

Q: What documents do I need for international travel?
A: For international travel, you typically need: 1) A valid passport (valid for at least 6 months), 2) Visa (if required for your destination), 3) Travel insurance, 4) Vaccination certificates (if required), 5) Return ticket.

Q: How do I get a visa?
A: Visa requirements depend on your destination and nationality. Apply through the embassy/consulate of your destination country...
```

````python
import lancedb

db = lancedb.connect("./lancedb")
faq_table = db.open_table("faq_table")
print("FAQ table opened. Rows:", faq_table.count_rows())
````

````python
def extract_qa(record: dict):
    for qk, ak in [("question", "answer"), ("q", "a"), ("Question", "Answer")]:
        if qk in record and ak in record:
            return str(record[qk]).strip(), str(record[ak]).strip()

    text = str(record.get("text", "")).strip()
    if ("question" in text.lower() and "answer" in text.lower()) or (text.startswith("{") and text.endswith("}")):
        import re
        mq = re.search(r"(question|q)\s*[:=]\s*['\"]?(.*?)['\"]?\s*(,|\n)", text, flags=re.I)
        ma = re.search(r"(answer|a)\s*[:=]\s*['\"]?(.*)", text, flags=re.I)
        if mq and ma:
            q = mq.group(2).strip()
            a = ma.group(2).strip().strip("}").strip()
            return q, a

    return None, text
````

````python
from langchain_core.tools import tool

@tool
def faq_search(query: str):
    """
    Search the FAQ knowledge base using semantic vector search (LanceDB + embeddings).

    Input:
      - query: user's natural language question

    Output:
      - Top 2 FAQ matches formatted as:
        Based on our FAQ:
        Q: ...
        A: ...
    """
    if not query or not query.strip():
        return {"error": "Query is missing. Please ask a question."}

    try:
        qvec = embedding_model.encode(query.strip())

        results = (faq_table.search(qvec).limit(2).to_list())

        if not results:
            return "No relevant FAQ found."

        blocks = ["Based on our FAQ:"]
        for r in results:
            q, a = extract_qa(r)
            if not q:
                q = query.strip()

            if not a:
                a = "No answer text available in the FAQ record."

            blocks.append(f"OrgQuestion: {q} OrgAnswer: {a}")

        return "".join(blocks).strip()

    except Exception as e:
        return {"error": f"FAQ search error: {str(e)}"}
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Trip Planning</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

````python
import lancedb

db = lancedb.connect("./lancedb")
book_table = db.open_table("book_table")
print("Book table opened. Rows:", book_table.count_rows())
````

````python
from datetime import datetime, timedelta

def compute_trip_days(arrival_date: str, departure_date: str):
    a_norm, err = normalize_date(arrival_date)
    if err:
        return None, None, f"Arrival date error: {err}"

    d_norm, err = normalize_date(departure_date)
    if err:
        return None, None, f"Departure date error: {err}"

    a = datetime.strptime(a_norm, "%Y-%m-%d").date()
    d = datetime.strptime(d_norm, "%Y-%m-%d").date()

    if d < a:
        return None, None, "Departure date must be after arrival date."

    num_days = (d - a).days
    if num_days == 0:
        num_days = 1

    days = [a + timedelta(days=i) for i in range(num_days)]
    return a_norm, d_norm, days
````

````python
def clean_text(s: str) -> str:
    s = s or ""
    s = s.replace("\n", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s
````

````python
def build_trip_queries(destination: str, interests: list[str] | None):
    dest = destination.strip()
    base = [
        f"top attractions in {dest}",
        f"must see places in {dest}",
        f"local food and cuisine in {dest}",
        f"best areas to stay in {dest}",
        f"things to do in {dest}",
    ]
    if interests:
        for it in interests:
            it = it.strip().lower()
            if it:
                base.append(f"{it} experiences in {dest}")
    return base
````

````python
def rag_search_book(queries: list[str], per_query_k: int = 5):
    all_hits = []
    seen = set()

    for q in queries:
        qvec = embedding_model.encode(q)
        hits = book_table.search(qvec).limit(per_query_k).to_list()

        for h in hits:
            key = (h.get("doc_id"), h.get("chunk_index"))
            if key in seen:
                continue
            seen.add(key)
            all_hits.append(h)

    return all_hits
````

````python
def tavily_fallback(destination: str, interests: list[str] | None, max_results: int = 5):
    qs = [
        f"{destination} must see attractions",
        f"{destination} local food dishes",
        f"{destination} best neighborhoods to stay",
    ]
    if interests:
        qs.append(f"{destination} {' '.join(interests)} things to do")

    merged = ""
    urls = []
    for q in qs:
        res = tavily_client.search(q, max_results=2)
        for r in res.get("results", []):
            merged += "\n" + (r.get("content") or r.get("snippet") or "")
            if r.get("url"):
                urls.append(r["url"])
    return merged.strip(), list(dict.fromkeys(urls))[:max_results]
````

````python
import re

PERSON_BLACKLIST = {
    "Henry Miller", "Oscar Wilde", "Raquel Carena"
}

def extract_place_candidates(text: str, destination: str, max_items: int = 25):
    text = clean_text(text)

    candidates = re.findall(r"\b([A-Z][a-z]+(?:\s+[A-Z][a-z]+){1,4})\b", text)

    out, seen = [], set()
    for c in candidates:
        c = clean_text(c)

        if c.lower() == destination.lower():
            continue

        if c in PERSON_BLACKLIST:
            continue

        if any(bad in c.lower() for bad in ["world travel book", "please don", "me i", "now here"]):
            continue

        if len(c) < 5:
            continue

        key = c.lower()
        if key not in seen:
            seen.add(key)
            out.append(c)

        if len(out) >= max_items:
            break

    return out
````

````python
def get_restaurant_recs(destination: str, max_results: int = 5):
    try:
        res = restaurant_search.invoke({"destination": destination, "max_results": max_results})
        if isinstance(res, dict) and res.get("results"):
            recs = res["results"]
            recs = [r for r in recs if r.get("name") and len(r["name"]) > 2]
            return recs[:max_results], None
        return [], None
    except Exception as e:
        return [], f"Restaurant search error: {str(e)}"
````

````python
def build_itinerary(days, destination: str, interests: list[str] | None, context_text: str, restaurants: list[dict] | None = None):
    interests = interests or []
    interests_txt = ", ".join(interests) if interests else "general sightseeing"

    places = extract_place_candidates(context_text, destination, max_items=30)

    priority = []
    if destination.lower() == "paris":
        for p in ["Eiffel Tower", "Notre-Dame", "Arc de Triomphe", "Louvre", "Seine", "Montmartre"]:
            priority.append(p)

    for p in places:
        if p not in priority:
            priority.append(p)

    restaurants = restaurants or []

    itinerary = []
    idx = 0
    ridx = 0

    for i, day in enumerate(days, start=1):
        p1 = priority[idx % len(priority)] if priority else f"a major landmark in {destination}"
        p2 = priority[(idx+1) % len(priority)] if priority else f"a notable neighborhood in {destination}"
        p3 = priority[(idx+2) % len(priority)] if priority else f"a cultural spot in {destination}"
        idx += 3

        dinner_line = "Enjoy local dinner and a relaxed evening walk."
        if restaurants:
            r = restaurants[ridx % len(restaurants)]
            ridx += 1
            rname = r.get("name", "a top restaurant")
            rurl = r.get("url", "")
            rdesc = (r.get("description") or "").strip()

            if rurl:
                dinner_line = f"Dinner at {rname} — {rdesc[:120].rstrip()}... More: {rurl}"
            else:
                dinner_line = f"Dinner at {rname} — {rdesc[:120].rstrip()}..."

        itinerary.append({
            "day": i,
            "date": day.strftime("%Y-%m-%d"),
            "morning": f"Visit {p1} (focus: {interests_txt}).",
            "afternoon": f"Explore {p2} and try a local lunch nearby.",
            "evening": f"{dinner_line} Then relax around {p3}.",
            "notes": "Attractions are from the travel guide (RAG). Restaurants come from web search."
        })

    return itinerary
````

````python
from langchain_core.tools import tool

@tool
def trip_planning(destination: str, arrival_date: str, departure_date: str, interests: str = ""):
    """
    Create a daily travel itinerary using RAG (LanceDB book_table). Falls back to Tavily if info is insufficient.

    Inputs:
      - destination: city/country (required)
      - arrival_date: date in various formats
      - departure_date: date in various formats
      - interests: comma-separated interests (e.g., "food, culture, shopping")

    Output:
      - daily itinerary + sources
    """
    if not destination or not destination.strip():
        return {"error": "Destination is missing."}

    a_norm, d_norm, days_or_err = compute_trip_days(arrival_date, departure_date)
    if isinstance(days_or_err, str):
        return {"error": days_or_err}

    days = days_or_err

    interests_list = [x.strip() for x in interests.split(",") if x.strip()] if interests else []

    queries = build_trip_queries(destination, interests_list)

    try:
        hits = rag_search_book(queries, per_query_k=5)
    except Exception as e:
        return {"error": f"Book RAG search error: {str(e)}"}

    context_chunks = [h.get("text", "") for h in hits if h.get("text")]
    context_text = "\n\n".join(context_chunks[:15]).strip()

    sources = []
    if hits:
        sources = [
            {"source": h.get("source"), "doc_id": h.get("doc_id"), "chunk_index": h.get("chunk_index")}
            for h in hits[:6]
        ]

    tavily_text = ""
    tavily_sources = []
    if len(context_text.split()) < 120:
        tavily_text, tavily_sources = tavily_fallback(destination, interests_list)
    
    restaurants, rerr = get_restaurant_recs(destination, max_results=4)
    itinerary = build_itinerary(days, destination, interests_list, context_text, restaurants=restaurants)
    
    return {
        "destination": destination.strip(),
        "arrival_date": a_norm,
        "departure_date": d_norm,
        "num_days": len(days),
        "interests": interests_list,
        "itinerary": itinerary,
        "restaurant_recommendations": [
            {"name": r.get("name"), "url": r.get("url"), "cuisine_type": r.get("cuisine_type")}
            for r in restaurants[:4]],
        "rag_context_preview": context_text[:800],
        "rag_sources": sources,
        "web_fallback_used": bool(tavily_text),
        "web_sources": tavily_sources
    }
````

````python
from typing import Annotated, TypedDict
from langchain_core.messages import BaseMessage
from langgraph.graph.message import add_messages

class GraphState(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages]
````

````python
from langchain_core.messages import SystemMessage

def reasoner_node(state: GraphState) -> GraphState:
    history = state.get("messages", [])
    messages = [SystemMessage(content=SYSTEM_MESSAGE)] + history
    response = llm_with_tools.invoke(messages)
    return {"messages": [response]}
````

````python
from langchain_core.messages import HumanMessage

test_state = {"messages": [HumanMessage(content="Find 3 flights from Tehran to Tokyo on Feb 10 2026.")]}

out = reasoner_node(test_state)

print("Reasoner node output type:", type(out))
print("Messages in state:", len(out["messages"]))
print("Last message:", out["messages"][-1])
````

````python
from langgraph.graph import StateGraph, END

builder = StateGraph(GraphState)
builder.add_node("reasoner", reasoner_node)
builder.add_node("tools", tools_node)

print("Nodes added to graph.")
````

````python
from langgraph.prebuilt import tools_condition

builder.set_entry_point("reasoner")
builder.add_conditional_edges(
    "reasoner",
    tools_condition,
    {
        "tools": "tools",
        END: END
    }
)

builder.add_edge("tools", "reasoner")
print("Edges added.")
````

````python
from langchain_core.messages import HumanMessage, AIMessage

def _extract_tool_names(messages):
    tool_names = []
    for m in messages:
        if isinstance(m, AIMessage) and getattr(m, "tool_calls", None):
            for tc in m.tool_calls:
                name = tc.get("name")
                if name:
                    tool_names.append(name)

    seen, uniq = set(), []
    for n in tool_names:
        if n not in seen:
            uniq.append(n)
            seen.add(n)
    return uniq


def _extract_final_answer(messages):
    for m in reversed(messages):
        if isinstance(m, AIMessage):
            content = (m.content or "").strip()
            if content:
                return content
    return "(No final text response produced.)"


def chat_cli(graph, max_turns: int = 10):
    print("TravelBot CLI is ready.")
    print(f"Type your message. (max turns = {max_turns})")
    print("Commands: 'exit' to quit, 'clear' to reset conversation.\n")

    state = {"messages": []}
    turns = 0

    while turns < max_turns:
        user_input = input(f"You ({turns+1}/{max_turns}): ").strip()

        if not user_input:
            continue

        cmd = user_input.lower()
        if cmd in ["exit", "quit"]:
            print("Bye")
            return

        if cmd == "clear":
            state = {"messages": []}
            print("Conversation cleared.\n")
            continue

        state["messages"].append(HumanMessage(content=user_input))

        try:
            state = graph.invoke(state)

            used_tools = _extract_tool_names(state["messages"])
            if used_tools:
                print(f"Tools used: {', '.join(used_tools)}")

            answer = _extract_final_answer(state["messages"])
            print(f"Bot: {answer}\n")

        except Exception as e:
            print(f"Error: {str(e)}\n")

        turns += 1

    print("Max turns reached. CLI session ended.")
````

````python
from langchain_core.messages import HumanMessage, AIMessage

def chat_cli_once(graph, user_input: str, state=None):
    """
    Run ONE CLI turn safely (no loop, no input()).
    Useful for testing in Jupyter without kernel blocking.
    """
    print("You:", user_input)

    if state is None:
        state = {"messages": []}

    state["messages"].append(HumanMessage(content=user_input))

    state = graph.invoke(state)

    used_tools = _extract_tool_names(state["messages"])
    if used_tools:
        print("Tools used:", ", ".join(used_tools))
    else:
        print("Tools used: none")

    answer = _extract_final_answer(state["messages"])
    print("Bot:", answer)
    print("=======================\n")

    return state
````

در این سناریو، عامل از ابزار <b>hotel_search</b> برای یافتن هتل در دبی با تاریخ‌های مشخص استفاده کرد. خروجی شامل چند هتل واقعی با قیمت هر شب، مجموع هزینه اقامت و واحد پول بود. اگرچه برخی فیلدها مانند rating یا آدرس در داده‌های آمادئوس موجود نبودند، عامل این موضوع را شفاف بیان کرد و از جعل اطلاعات خودداری نمود. کیفیت پاسخ خوب ارزیابی می‌شود زیرا اطلاعات اصلی اقامت به‌درستی ارائه شده است.<br>
<br>

<b>سناریو ۴ (جستجوی هتل – پاریس با محدودیت بودجه):</b><br>
در این سناریو ابزار <b>hotel_search</b> ابتدا تلاش کرد داده‌های زنده آمادئوس را بازیابی کند، اما نتیجه‌ای در بازه بودجه مشخص‌شده یافت نشد. عامل سپس به‌درستی از fallback وب استفاده کرد و صفحات معتبر راهنمای هتل را معرفی نمود، ضمن اینکه صراحتاً اعلام کرد این‌ها پیشنهادهای لیستی هستند نه قیمت‌های زنده. این شفافیت و پیشنهاد گام بعدی (تغییر بودجه یا فیلتر) نشان‌دهنده کیفیت بالای پاسخ و مدیریت مناسب محدودیت داده‌هاست.<br>
<br>

<b>سناریو ۵ (جستجوی رستوران – پاریس):</b><br>
در این سناریو ابزار <b>restaurant_search</b> فراخوانی شد و پنج رستوران واقعی با توضیح کوتاه و لینک ارائه گردید. نتایج از منابع مختلف وب انتخاب شده و توضیحات هر رستوران اختصاصی و غیرتکراری بود. کیفیت پاسخ بالا است زیرا تنوع رستوران‌ها، شفافیت منبع و ارتباط مستقیم توضیح با همان رستوران رعایت شده است.<br>
<br>

<b>سناریو ۶ (جستجوی رستوران – دبی):</b><br>
عامل با استفاده از <b>restaurant_search</b> ترکیبی از رستوران‌های محلی و محبوب دبی را ارائه داد و آن‌ها را از نظر سبک غذایی و محبوبیت توضیح داد. وجود لینک مستقیم برای هر مورد و تنوع منابع باعث می‌شود این پاسخ برای کاربر عملی و قابل استفاده باشد. کیفیت پاسخ بسیار مناسب ارزیابی می‌شود.<br>
<br>

<b>سناریو ۷ (اطلاعات آب‌وهوا – دبی):</b><br>
در این سناریو ابزار <b>get_weather_info</b> برای تاریخ مشخص فراخوانی شد و اطلاعاتی شامل دما، وضعیت هوا و توصیه پوشش ارائه گردید. عامل همچنین منابع وب مورد استفاده را ذکر کرد. اگرچه برخی پارامترها مانند رطوبت یا احتمال بارش در منابع یافت نشدند، این موضوع صادقانه اعلام شد. کیفیت پاسخ بالا و مطابق خواسته پروژه است.<br>
<br>

<b>سناریو ۸ (اطلاعات آب‌وهوا – پاریس):</b><br>
عامل با نرمال‌سازی تاریخ و استفاده از <b>get_weather_info</b> پیش‌بینی هوا، دما (با تبدیل واحد)، و توصیه پوشش را ارائه داد. ذکر منابع متعدد و توضیح عملی برای لباس مناسب، کیفیت پاسخ را بالا می‌برد. این سناریو نشان می‌دهد ابزار آب‌وهوا به‌درستی با فرمت‌های مختلف تاریخ کار می‌کند.<br>
<br>

<b>سناریو ۹ (تبدیل ارز – USD به AED):</b><br>
در این سناریو ابزار <b>get_currency_info</b> با کدهای ارزی مستقیم استفاده شد و نرخ تبدیل، نرخ معکوس و منابع معتبر وب ارائه گردید. خروجی شفاف، دقیق و همراه با توضیح درباره نوسان نرخ‌هاست. کیفیت پاسخ بسیار خوب و حرفه‌ای است.<br>
<br>

<b>سناریو ۱۰ (تبدیل ارز – ایران به فرانسه):</b><br>
عامل با map کردن نام کشورها به کدهای ISO (IRR و EUR) ابزار <b>get_currency_info</b> را فراخوانی کرد. نرخ منطقی، مثال عددی (تبدیل 1,000,000 ریال) و منابع معتبر ارائه شد. این سناریو نشان‌دهنده صحت mapping و جلوگیری از خطاهای رایج در نرخ ارز است و کیفیت پاسخ عالی است.<br>
<br>

<b>سناریو ۱۱ (سؤال FAQ – زمان حضور در فرودگاه):</b><br>
در این سناریو ابزار <b>faq_search</b> مطابق قوانین routing فراخوانی شد و پاسخ به‌صورت Q/A از پایگاه FAQ ارائه گردید. پاسخ دقیق، مختصر و کاملاً مطابق سیاست‌های رایج سفر هوایی است. کیفیت پاسخ بسیار خوب است زیرا از دانش RAG استفاده شده و نه دانش عمومی مدل.<br>
<br>

<b>سناریو ۱۲ (سؤال FAQ – بردن غذا در هواپیما):</b><br>
عامل با استفاده از <b>faq_search</b> دو پاسخ مرتبط (قوانین غذا و اقلام ممنوعه) را برگرداند. رعایت limit(2) و فرمت استاندارد Q/A نشان‌دهنده پیاده‌سازی صحیح ابزار است. کیفیت پاسخ بالا و کاربردی ارزیابی می‌شود.<br>
<br>

<b>سناریو ۱۳ (برنامه‌ریزی سفر – پاریس ۴ روزه):</b><br>
در این سناریو ابزار <b>trip_planning</b> به‌تنهایی استفاده شد و برنامه‌ای روزبه‌روز مبتنی بر RAG از کتاب سفر تولید گردید. ساختار morning/afternoon/evening، لحاظ کردن علایق (food و culture) و ارائه توضیحات عملی نشان‌دهنده کیفیت بالای پاسخ است. همچنین اشاره به منابع RAG باعث افزایش اعتبار خروجی شده است.<br>
<br>

<b>سناریو ۱۴ (برنامه‌ریزی سفر ترکیبی – پاریس ۳ روزه با رستوران):</b><br>
عامل از ترکیب <b>trip_planning</b> و <b>restaurant_search</b> استفاده کرد و برنامه‌ای سه‌روزه همراه با پیشنهاد شام برای هر شب ارائه داد. این سناریو قدرت معماری ReAct و استفاده ترکیبی از ابزارها را نشان می‌دهد. کیفیت پاسخ بسیار بالا است زیرا خروجی هم غنی، هم عملی و هم مستند به منابع مختلف است.<br>
<br>

<b>نتیجه‌گیری کلی:</b><br>
نتایج تست‌ها نشان می‌دهد سیستم TravelBot از نظر انتخاب ابزار، مدیریت خطا، استفاده از RAG و ارائه پاسخ‌های کاربرپسند عملکرد بسیار خوبی دارد. نقاط قوت اصلی سیستم شامل routing دقیق ابزارها، عدم تولید اطلاعات جعلی، پاسخ‌های ساختاریافته و امکان ترکیب چند ابزار در یک سناریو است. از نقاط قابل بهبود می‌توان به غنی‌تر کردن اطلاعات هتل (مانند آدرس و rating در صورت دسترسی) و پالایش بیشتر نتایج رستوران‌های لیستی اشاره کرد. در مجموع، سیستم پیاده‌سازی‌شده کاملاً با اهداف پروژه هم‌راستا بوده و آماده استفاده و توسعه در مقیاس بزرگ‌تر است.
