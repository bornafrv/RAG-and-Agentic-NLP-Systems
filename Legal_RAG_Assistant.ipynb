```python
import fitz  # PyMuPDF

def is_scanned_pdf(pdf_path, min_chars=100):

    doc = fitz.open(pdf_path)
    total_text_len = 0

    for page in doc:
        text = page.get_text("text")
        if text:
            total_text_len += len(text.strip())

    return total_text_len < min_chars

```

```python
import fitz
import os

def pdf_to_images(pdf_path, out_dir, dpi=400):
    doc = fitz.open(pdf_path)
    image_paths = []

    base = os.path.splitext(os.path.basename(pdf_path))[0]
    pdf_out_dir = os.path.join(out_dir, base)
    os.makedirs(pdf_out_dir, exist_ok=True)

    for i, page in enumerate(doc):
        pix = page.get_pixmap(dpi=dpi)
        img_path = os.path.join(pdf_out_dir, f"page_{i+1:04d}.png")
        pix.save(img_path)
        image_paths.append(img_path)

    return image_paths
```

```python
from vllm import LLM, SamplingParams


llm = LLM(
    model="deepseek-ai/DeepSeek-OCR",
    trust_remote_code=True,
    dtype="float16",
    max_model_len=4096,
    enforce_eager=True
)

sampling_params = SamplingParams(temperature=0, max_tokens=2048)
```

```python
from PIL import Image, ImageEnhance, ImageOps, ImageFilter

def preprocess_for_ocr(img: Image.Image) -> Image.Image:
    img = img.convert("RGB")

    scale = 2
    img = img.resize((img.width * scale, img.height * scale), Image.BICUBIC)

    gray = ImageOps.grayscale(img)

    gray = ImageEnhance.Contrast(gray).enhance(2.8)

    gray = gray.filter(ImageFilter.SHARPEN)
    gray = ImageEnhance.Sharpness(gray).enhance(2.0)

    bw = gray.point(lambda x: 255 if x > 160 else 0)

    return bw.convert("RGB")
```

```python
from PIL import Image

def ocr_image_strong(image_path):
    img_raw = Image.open(image_path).convert("RGB")

    prompt = "<image>\nExtract the Persian text exactly as seen."

    out1 = llm.generate(
        {"prompt": prompt, "multi_modal_data": {"image": img_raw}},
        sampling_params=sampling_params,
        use_tqdm=False
    )[0].outputs[0].text.strip()

    if len(out1) > 200:
        return out1

    img_proc = preprocess_for_ocr(img_raw)
    out2 = llm.generate(
        {"prompt": prompt, "multi_modal_data": {"image": img_proc}},
        sampling_params=sampling_params,
        use_tqdm=False
    )[0].outputs[0].text.strip()

    return out2
```

```python
def ocr_pdf_strong(pdf_path, image_root_dir, out_txt_path, dpi=500):

    image_paths = pdf_to_images(pdf_path, image_root_dir, dpi=dpi)

    all_text = []
    total = len(image_paths)

    print(f"\nOCR started for: {os.path.basename(pdf_path)}")
    print(f"Total pages: {total}")

    for i, img_path in enumerate(image_paths, start=1):
        print(f"  → Page {i}/{total}")

        raw_text = ocr_image_strong(img_path)

        all_text.append(f"\n===== PAGE {i} =====\n{raw_text}")

    final_text = "\n".join(all_text)

    with open(out_txt_path, "w", encoding="utf-8") as f:
        f.write(final_text)

    return out_txt_path
```

```python
import re
import pandas as pd
import datetime

ARTICLE_RE = re.compile(r"(?m)^\s*ماده\s+([۰-۹0-9]+(?:\s*مکرر(?:\s*[۰-۹0-9]+)?)?)\b")

def fa_to_en_digits(s: str) -> str:
    return s.translate(str.maketrans("۰۱۲۳۴۵۶۷۸۹", "0123456789"))

def extract_article_number(article_title: str) -> str:
    x = article_title.replace("ماده", "").strip()
    return fa_to_en_digits(x)

def chunk_by_article_with_metadata(text: str, law_title: str, law_domain: str):
    text = text.replace("\r", "")
    matches = list(ARTICLE_RE.finditer(text))

    chunks = []
    for i, m in enumerate(matches):
        start = m.start()
        end = matches[i + 1].start() if i + 1 < len(matches) else len(text)

        article_title = m.group(1)
        article_no = fa_to_en_digits(article_title)
        chunk_text = text[start:end].strip()

        has_tabserah = "تبصره" in chunk_text

        chunks.append({
            "law_title": law_title,
            "law_domain": law_domain,
            "article_no": article_no,
            "text": chunk_text,
            "has_tabserah": has_tabserah,
            "char_len": len(chunk_text),
            "page_start": start,
            "page_end": end
        })

    return chunks


def guess_domain_from_name(filename: str) -> str:
      name = filename.replace("_", " ").lower()
      if "كار" in filename or "کار" in filename: return "labor"
      if "تأمين" in filename or "تامین" in filename: return "social_security"
      if "چك" in filename or "چک" in filename: return "check"
      if "موجر" in filename or "مستأجر" in filename or "مستاجر" in filename: return "landlord_tenant"
      if "ماليات" in filename or "مالیات" in filename: return "vat_tax"
      if "قاچاق" in filename: return "anti_smuggling"
      if "ثبت اسناد" in filename or "املاك" in filename or "املاک" in filename: return "registration_realestate"
      return "other"
```

```python
import os
import glob

TXT_DIR = "/content/law_dataset/law_dataset"
txt_paths = sorted(glob.glob(os.path.join(TXT_DIR, "*.txt")))

all_chunks = []

for path in txt_paths:
    law_title = os.path.splitext(os.path.basename(path))[0]
    law_domain = guess_domain_from_name(law_title)

    with open(path, encoding="utf-8") as f:
        text = f.read()

    chunks = chunk_by_article_with_metadata(text, law_title, law_domain)

    all_chunks.extend(chunks)

df = pd.DataFrame(all_chunks)
df.head(), df.shape
```

```python
from transformers import AutoTokenizer, AutoModel
import torch
import numpy as np

model_name = "intfloat/multilingual-e5-large"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
model.eval()
```

```python
def embed_text(text: str):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=512
    )
    with torch.no_grad():
        outputs = model(**inputs)
        embedding = outputs.last_hidden_state.mean(dim=1)
    return embedding.squeeze().cpu().numpy()

def embed_chunk(chunk_text: str):
    return embed_text("passage: " + chunk_text)

def embed_query(query: str):
    return embed_text("query: " + query)
```

```python
def apply_embedding_to_chunks(df):
    embeddings = []
    for _, row in df.iterrows():
        emb = embed_chunk(row["text"])
        embeddings.append(emb)
    return embeddings

df["embedding"] = apply_embedding_to_chunks(df)
```

```python
import lancedb
import pyarrow as pa
import numpy as np

EMBEDDING_DIM = 1024

schema = pa.schema([
    pa.field("law_title", pa.string()),
    pa.field("law_domain", pa.string()),
    pa.field("article_no", pa.string()),
    pa.field("text", pa.string()),
    pa.field("has_tabserah", pa.bool_()),
    pa.field("char_len", pa.int64()),
    pa.field(
        "embedding",
        pa.list_(pa.float32(), EMBEDDING_DIM)
    )
])

```

```python
arrow_table = pa.Table.from_pandas(
    df,
    schema=schema,
    preserve_index=False
)
```

```python
import lancedb

db = lancedb.connect("/content/lancedb_storage")
TABLE_NAME = "law_chunks"

if TABLE_NAME in db.table_names():
    db.drop_table(TABLE_NAME)
    print("Old table dropped")

table = db.create_table(TABLE_NAME, arrow_table)
```

```python
import numpy as np

query_text = "شرایط بازنشستگی بیمه شده چیست؟"
query_embedding = np.asarray(embed_query(query_text), dtype=np.float32)

results = (
    table
    .search(query_embedding, vector_column_name="embedding")
    .limit(5)
    .to_list()
)

for i, r in enumerate(results, 1):
    print("="*80)
    print("Rank:", i)
    print("law_title:", r.get("law_title"))
    print("article_no:", r.get("article_no"))
    print("has_tabserah:", r.get("has_tabserah"))
    print("text:", r.get("text","")[:500])
```

قابلیت های کلی Agent باید به صورت زیر باشد :

یک نود با نام rewrite_query طراحی کنید که ورودی آن، query اصلی کاربر و خروجی آن، یک rewritten_query به‌عنوان نسخه‌ای بازنویسی‌شده و شفاف‌تر از سؤال باشد. وظیفه این ماژول آن است که ابهام‌های زبانی موجود در پرسش را برطرف کند، لحن را استاندارد و یکنواخت سازد و در نهایت پرسش کاربر را به فرمی تبدیل کند که برای مرحله‌ی بازیابی اطلاعات (retrieval) مناسب‌تر باشد.



نودی با نام classify_intent طراحی کنید که وظیفه آن تشخیص نوع پرسش کاربر بر اساس query دریافتی است. این نود باید تعیین کند که پرسش کاربر در کدام یک از دسته‌های «Greeting / Small Talk» (مانند سلام، احوال‌پرسی، تبریک و عبارات مشابه)، «Abusive / Insult / بی‌احترامی» یا «Law Question» (سؤال مرتبط با قانون و نیازمند مراجعه به پایگاه دانش) قرار می‌گیرد. در صورت تشخیص دسته‌ی Greeting، نود باید یک پاسخ کوتاه، محترمانه و دوستانه تولید کند. در صورت تشخیص دسته‌ی Abusive، نود موظف است پاسخی حرفه‌ای، محترمانه و بدون هرگونه توهین ارائه دهد و در آن توضیح دهد که مدل برای کمک طراحی شده است و وارد جدل یا بحث تنش‌زا نمی‌شود. اگر خروجی نود در دسته‌ی Law Question قرار گرفت، جریان پردازش باید به نود بعدی، یعنی نود «Metadata Extraction»، منتقل شود تا مراحل بعدی بازیابی اطلاعات انجام گیرد.





یک نود با نام extract_metadata طراحی کنید که ورودی آن rewritten_query و خروجی آن یک ساختار metadata_filter باشد؛ این ساختار در مرحله‌ی بازیابی اطلاعات از LanceDB برای اعمال فیلترهای مرتبط استفاده خواهد شد. در این نود، با استفاده از یک LLM، از روی rewritten_query اطلاعات ساختارمند و قابل‌استفاده به‌عنوان فیلتر استخراج می‌شود تا در قالب metadata_filter به LanceDB ارسال شده و بازیابی اسناد با دقت و ارتباط بیشتری انجام پذیرد.



نودی با نام context_retrieve طراحی کنید که با استفاده از rewritten_query در LanceDB عملیات جستجو را انجام دهد. این نود باید از ساختار metadata_filter برای محدود کردن فضای جستجو و اعمال فیلترهای موردنظر بهره بگیرد و در نهایت، تعداد مشخصی سند (chunk) مرتبط با پرسش کاربر، مثلاً به تعداد K سند (برای نمونه K = 10)، را به‌عنوان خروجی بازگرداند.



یک نود با نام rerank طراحی کنید که روی خروجی مرحله‌ی context_retrieve یک لایه‌ی بازمرتب‌سازی (Reranking) اعمال کند؛ به‌عنوان مثال می‌توان از یک LLM-based reranker یا یک معیار شباهت ثانویه برای امتیازدهی مجدد به اسناد استفاده کرد. این نود باید بر اساس مقادیر similarity یا score، تعداد محدودی از بهترین سندها (top-N، مثلاً N = 3) را برای مرحله‌ی تولید پاسخ (generation) انتخاب کند. علاوه بر این، در همین نود باید یک مرحله‌ی بررسی ارتباط (Relevance Check) انجام شود؛ به این صورت که اگر هیچ‌یک از contextهای انتخاب‌شده در top-N، بر اساس یک آستانه‌ی تعریف‌شده یا قضاوت LLM، ارتباط کافی با query نداشتند، فرآیند retrieval یک‌بار دیگر تکرار شده و تلاش مجدد برای بازیابی contextهای مرتبط‌تر انجام گیرد. دقت شود که نهایتا یک بار می بایست این مرحله انجام شود و Agent در loop بی نهایت قرار نگیرد پس این موضوع باید توسط شما کنترل شود.





نودی با نام generate_answer طراحی کنید که ورودی آن شامل rewritten_query به‌همراه contextهای نهایی انتخاب‌شده از مراحل قبل باشد و بر اساس این ورودی، وظیفه دارد یک پاسخ متنی نهایی، منسجم، دقیق و قابل‌فهم برای کاربر تولید و به‌عنوان خروجی برگرداند.


به طور کلی معماری پیشنهادی یک حالت از نحوه پیاده سازی این Agent هست و می توان نوع نگاه متفاوتی در طراحی معماری Agent داشت پس شما مجاز هستید که با پیاده سازی همه مواردی که بیان شد با معماری متفاوت عمل کنید معماری پیشنهادی شما می بایست تمامی این موارد را شامل شده باشد:

query rewriting, extract metadata , context retrieve , reranking , generate answer

```python
from fastembed import TextEmbedding

embedder = TextEmbedding(model_name=EMB_MODEL_NAME)

def embed_query(text: str) -> list[float]:
    return list(embedder.embed([f"query: {text}"]))[0].tolist()

def embed_passage(text: str) -> list[float]:
    return list(embedder.embed([f"passage: {text}"]))[0].tolist()

print("FastEmbed ready. Dim =", len(embed_query("تست")))
```

```python
from typing import TypedDict, List, Dict, Any

class AgentState(TypedDict, total=False):
    user_query: str
    rewritten_query: str

    intent: str
    intent_response: str

    metadata_filter: Dict[str, Any]

    retrieved_docs: List[Dict[str, Any]]
    reranked_docs: List[Dict[str, Any]]

    retry_count: int
    need_retry: bool

    final_answer: str

    timings: List[Dict[str, Any]]
    question_id: str
    source_law: str
```

```python
from langchain_openai import ChatOpenAI
import os

def get_llm():

    return ChatOpenAI(
        model="gpt-5-mini",
        temperature=0,
        api_key=os.getenv("OPENAI_API_KEY"),
        base_url=os.getenv("OPENAI_BASE_URL"),
        # api_key=os.getenv("aa-zT3t9Ssf7sup3wgBp6Fb7heNy8NkjT7tLQQHPr8hvp8ugn6e"),
        # base_url=os.getenv("https://api.avalai.ir/v1"),
    )
```

```python
from langchain_core.prompts import ChatPromptTemplate

rewrite_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "تو فقط وظیفه بازنویسی پرسش کاربر را داری.\n"
        "سؤال را شفاف، دقیق و مناسب بازیابی اسناد قانونی کن.\n"
        "از خودت پاسخ نده، توضیح نده، فقط متن سؤال بازنویسی‌شده را برگردان."
    ),
    ("human", "{question}")
])


def rewrite_query_node(state: AgentState) -> AgentState:
    user_q = state["user_query"]

    llm = get_llm()

    rewritten = llm.invoke(
        rewrite_prompt.format_messages(question=user_q)
    ).content.strip()

    return {
        "rewritten_query": rewritten
    }
```

```python
intent_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "نوع پیام کاربر را دقیقاً در یکی از این سه دسته تشخیص بده.\n"
        "فقط یکی از این سه کلمه را خروجی بده و هیچ توضیح اضافه نده:\n"
        "Greeting\nAbusive\nLaw"
    ),
    ("human", "{question}")
])


def classify_intent_node(state: AgentState) -> AgentState:
    user_q = state["user_query"]

    llm = get_llm()
    label = llm.invoke(
        intent_prompt.format_messages(question=user_q)
    ).content.strip()

    if label == "Greeting":
        return {
            "intent": "greeting",
            "intent_response": "سلام! خوشحال می‌شم سوال حقوقی‌ات رو بپرسی."
        }

    if label == "Abusive":
        return {
            "intent": "abusive",
            "intent_response": (
                "من برای کمک طراحی شدم و وارد گفت‌وگوی توهین‌آمیز نمی‌شوم. "
                "اگر سوالت را محترمانه مطرح کنی، با کمال میل راهنمایی می‌کنم."
            )
        }

    return {
        "intent": "law"
    }
```

```python
from typing import Literal, Optional

LAW_DOMAINS = Literal[
    "anti_smuggling",
    "check",
    "labor",
    "landlord_tenant",
    "registration_realestate",
    "social_security",
    "vat_tax"
]

LAW_TITLES = Literal[
    "قانون تأمين اجتماعي",
    "قانون ثبت اسناد و املاك",
    "قانون روابط موجر و مستأجر",
    "قانون صدور چك",
    "قانون كار",
    "قانون ماليات بر ارزش افزوده",
    "قانون مبارزه با قاچاق كالا و ارز"
]
```

```python
from pydantic import BaseModel, Field
from typing import Optional, Literal

class MetadataFilter(BaseModel):
    law_title: Optional[LAW_TITLES] = Field(
        default=None,
        description="نام قانون، فقط یکی از قوانین موجود در پایگاه داده"
    )

    law_domain: Optional[LAW_DOMAINS] = Field(
        default=None,
        description="دامنه موضوعی قانون، فقط از بین مقادیر مجاز دیتابیس"
    )

    article_no: Optional[str] = Field(
        default=None,
        description="شماره ماده در صورت اشاره مستقیم"
    )

    has_tabserah: Optional[bool] = Field(
        default=None,
        description="اگر سوال به تبصره اشاره دارد True"
    )

metadata_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "تو باید فقط از بین مقادیر مجاز زیر انتخاب کنی.\n\n"
        "دامنه‌های مجاز:\n"
        "- anti_smuggling\n"
        "- check\n"
        "- labor\n"
        "- landlord_tenant\n"
        "- registration_realestate\n"
        "- social_security\n"
        "- vat_tax\n\n"
        "عناوین قوانین مجاز:\n"
        "- قانون تأمين اجتماعي\n"
        "- قانون ثبت اسناد و املاك\n"
        "- قانون روابط موجر و مستأجر\n"
        "- قانون صدور چك\n"
        "- قانون كار\n"
        "- قانون ماليات بر ارزش افزوده\n"
        "- قانون مبارزه با قاچاق كالا و ارز\n\n"
        "اگر مطمئن نیستی، مقدار را null بگذار.\n"
        "حدس نزن."
    ),
    ("human", "{question}")
])


def extract_metadata_node(state: AgentState) -> AgentState:
    rewritten_q = state["rewritten_query"]

    llm = get_llm()

    extractor = llm.with_structured_output(MetadataFilter)

    metadata: MetadataFilter = extractor.invoke(
        metadata_prompt.format_messages(question=rewritten_q)
    )

    return {
        "metadata_filter": metadata.model_dump(exclude_none=True)
    }
```

```python
def build_where_clause(metadata_filter: dict) -> str:
    clauses = []

    if "law_title" in metadata_filter:
        clauses.append(f"law_title = '{metadata_filter['law_title']}'")

    if "law_domain" in metadata_filter:
        clauses.append(f"law_domain = '{metadata_filter['law_domain']}'")

    if "article_no" in metadata_filter:
        clauses.append(f"article_no = '{metadata_filter['article_no']}'")

    if "has_tabserah" in metadata_filter:
        val = "true" if metadata_filter["has_tabserah"] else "false"
        clauses.append(f"has_tabserah = {val}")

    return " AND ".join(clauses)


def context_retrieve_node(state: AgentState, K: int = 10) -> AgentState:
    query = state["rewritten_query"]
    metadata_filter = state.get("metadata_filter", {})

    query_embedding = embed_query(query)

    where_clause = build_where_clause(metadata_filter)

    search = table.search(query_embedding)

    if where_clause:
        search = search.where(where_clause)

    results = search.limit(K).to_list()

    return {
        "retrieved_docs": results
    }
```

```python
def rerank_node(state: AgentState, N: int = 3, min_score: float = 0.15) -> AgentState:
    docs = state.get("retrieved_docs", [])
    retry_count = state.get("retry_count", 0)

    def get_score(doc):
        if "score" in doc:
            return float(doc["score"])
        if "_distance" in doc:
            return 1.0 / (1.0 + float(doc["_distance"]))
        return 0.0

    docs_sorted = sorted(docs, key=get_score, reverse=True)

    top_docs = docs_sorted[:N]

    best_score = get_score(top_docs[0]) if top_docs else 0.0
    relevance_ok = best_score >= min_score

    need_retry = (not relevance_ok) and (retry_count == 0)

    return {
        "reranked_docs": top_docs,
        "relevance_ok": relevance_ok,
        "need_retry": need_retry
    }
```

```python
answer_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "تو یک دستیار حقوقی هستی که فقط و فقط بر اساس متن قوانین ارائه‌شده پاسخ می‌دهد.\n"
        "اگر اطلاعات کافی در متن‌های داده‌شده وجود نداشت، صریحاً بگو «اطلاعات کافی در اسناد بازیابی‌شده وجود ندارد».\n"
        "از خودت قانون، ماده یا تبصره اختراع نکن.\n"
        "در صورت امکان، در پاسخ نام قانون و شماره ماده را ذکر کن."
    ),
    (
        "human",
        "سؤال:\n{question}\n\n"
        "متن‌های قانونی بازیابی‌شده:\n{context}\n\n"
        "پاسخ نهایی:"
    )
])


def format_context_for_answer(docs: list[dict]) -> str:
    parts = []

    for d in docs:
        law = d.get("law_title", "نامشخص")
        article = d.get("article_no", "نامشخص")
        text = d.get("text", "")

        parts.append(
            f"[قانون: {law} | ماده: {article}]\n{text}"
        )

    return "\n\n---\n\n".join(parts)


def generate_answer_node(state: AgentState) -> AgentState:
    question = state["rewritten_query"]
    docs = state.get("reranked_docs", [])

    context_text = format_context_for_answer(docs)

    llm = get_llm()

    answer = llm.invoke(
        answer_prompt.format_messages(
            question=question,
            context=context_text
        )
    ).content.strip()

    return {
        "final_answer": answer
    }
```

```python
from langgraph.graph import StateGraph, END


def end_intent_node(state: AgentState) -> AgentState:
    return {"final_answer": state.get("intent_response", "")}

def bump_retry_node(state: AgentState) -> AgentState:
    return {
        "retry_count": state.get("retry_count", 0) + 1,
        "need_retry": False
    }

def intent_router(state: AgentState) -> str:
    intent = state.get("intent", "law")
    if intent in ("greeting", "abusive"):
        return "end_intent"
    return "extract_metadata"


def retry_router(state: AgentState) -> str:
    if state.get("need_retry", False):
        return "retry"
    return "generate"


graph = StateGraph(AgentState)

graph.add_node("rewrite_query", rewrite_query_node)
graph.add_node("classify_intent", classify_intent_node)

graph.add_node("end_intent", end_intent_node)

graph.add_node("extract_metadata", extract_metadata_node)
graph.add_node("context_retrieve", context_retrieve_node)
graph.add_node("rerank", rerank_node)

graph.add_node("bump_retry", bump_retry_node)

graph.add_node("generate_answer", generate_answer_node)

# Entry
graph.set_entry_point("rewrite_query")

# rewrite -> classify
graph.add_edge("rewrite_query", "classify_intent")

# classify -> (end_intent | extract_metadata)
graph.add_conditional_edges(
    "classify_intent",
    intent_router,
    {
        "end_intent": "end_intent",
        "extract_metadata": "extract_metadata",
    }
)

# end_intent -> END
graph.add_edge("end_intent", END)

# law path
graph.add_edge("extract_metadata", "context_retrieve")
graph.add_edge("context_retrieve", "rerank")

# rerank -> (retry | generate)
graph.add_conditional_edges(
    "rerank",
    retry_router,
    {
        "retry": "bump_retry",
        "generate": "generate_answer",
    }
)

# bump_retry -> context_retrieve
graph.add_edge("bump_retry", "context_retrieve")

# generate -> END
graph.add_edge("generate_answer", END)

app = graph.compile()
print("Graph compiled")
```

   در اطراف هر نود اصلی سیستم یعنی rewrite, classify_intent, extract_metadata, context_retrieve, rerank و generate_answer، زمان شروع و پایان اجرا را ثبت کنید. برای مجموعه‌ای شامل ۳ سوال از هر داکیومنت، کل pipeline سیستم RAG را اجرا کرده و زمان اجرای هر نود را برای هر پرسش لاگ کنید.

مشخص کنید گلوگاه زمانی (bottleneck) سیستم کجاست.

پیشنهاد دهید برای کاهش latency چه بهبودهایی می‌توان انجام داد. . </p>

```python
import time
from typing import Callable

def timed(node_name: str, fn: Callable[[AgentState], AgentState]) -> Callable[[AgentState], AgentState]:
    def _wrapped(state: AgentState) -> AgentState:
        if "timings" not in state:
            state["timings"] = []

        start = time.perf_counter()
        out = fn(state)
        end = time.perf_counter()

        state["timings"].append({
            "node": node_name,
            "start": start,
            "end": end,
            "duration_ms": (end - start) * 1000.0
        })

        return out
    return _wrapped
```

```python
from langgraph.graph import StateGraph, END

rewrite_t = timed("rewrite_query", rewrite_query_node)
classify_t = timed("classify_intent", classify_intent_node)
extract_t  = timed("extract_metadata", extract_metadata_node)

retrieve_t = timed(
    "context_retrieve",
    lambda s: context_retrieve_node(s, K=10)
)

rerank_t = timed(
    "rerank",
    lambda s: rerank_node(s, N=3, min_score=0.15)
)

generate_t = timed("generate_answer", generate_answer_node)

end_intent_t = timed("end_intent", end_intent_node)
bump_retry_t = timed("bump_retry", bump_retry_node)

graph_t = StateGraph(AgentState)

graph_t.add_node("rewrite_query", rewrite_t)
graph_t.add_node("classify_intent", classify_t)
graph_t.add_node("end_intent", end_intent_t)

graph_t.add_node("extract_metadata", extract_t)
graph_t.add_node("context_retrieve", retrieve_t)
graph_t.add_node("rerank", rerank_t)

graph_t.add_node("bump_retry", bump_retry_t)
graph_t.add_node("generate_answer", generate_t)

graph_t.set_entry_point("rewrite_query")

graph_t.add_edge("rewrite_query", "classify_intent")

graph_t.add_conditional_edges(
    "classify_intent",
    intent_router,
    {
        "end_intent": "end_intent",
        "extract_metadata": "extract_metadata",
    }
)

graph_t.add_edge("end_intent", END)

graph_t.add_edge("extract_metadata", "context_retrieve")
graph_t.add_edge("context_retrieve", "rerank")

graph_t.add_conditional_edges(
    "rerank",
    retry_router,
    {
        "retry": "bump_retry",
        "generate": "generate_answer",
    }
)

graph_t.add_edge("bump_retry", "context_retrieve")
graph_t.add_edge("generate_answer", END)

app_timed = graph_t.compile()

print("Timed graph compiled")
```

در تحلیل خطا ها، ابتدا موارد با Answer Relevancy پایین را بررسی می‌ کنیم. دو سوال در دیتاست امتیاز Answer Relevancy = 0.000 دارند: یکی درباره‌ ی تمدید قرارداد اجاره در قانون روابط موجر و مستاجر و دیگری درباره‌ ی آثار ثبت رسمی ملک در قانون ثبت اسناد و املاک. نکته ‌ی مهم این است که در هر دو مورد، سیستم به‌ جای پاسخ محتوایی، خروجی‌ ای از جنس اطلاعات کافی در اسناد بازیابی‌ شده وجود ندارد تولید کرده است. این نوع پاسخ از دید کاربر می‌ تواند رفتار محافظه‌ کارانه و درست باشد، اما RAGAS در معیار Answer Relevancy معمولا چنین پاسخ‌ هایی را عدم پاسخ‌ گویی مستقیم تلقی می‌ کند و به همین دلیل امتیاز را تا صفر پایین می‌ آورد؛ یعنی اینجا بیشتر با محدودیت معیار خودکار مواجهیم تا الزاما اشتباه معنایی سیستم. راه‌ حل عملی برای جلوگیری از افت شدید relevancy در چنین مواردی این است که الگوی پاسخ محافظه‌ کارانه اصلاح شود: به‌جای جمله‌ ی کلی اطلاعات کافی نیست، سیستم باید (1) دقیقا همان سوال را با یک جمله‌ ی کوتاه بازتاب دهد، (2) مشخص کند دقیقا چه جز اطلاعاتی در context نیست، و (3) اگر بخشی از پاسخ از context قابل استخراج است همان بخش را بدهد و فقط قسمت ناموجود را نامشخص اعلام کند. این کار هم از نظر کاربر بهتر است، هم RAGAS معمولا آن را مرتبط با سوال تشخیص می‌ دهد.

حالا موارد با Faithfulness پایین را علت‌ یابی می‌ کنیم. کمترین Faithfulness در نتایج مربوط به سوال تعریف قاچاق کالا و ارز در قانون مبارزه با قاچاق کالا و ارز با Faithfulness = 0.667 است. در این نمونه، پاسخ علاوه بر نقل تعریف، یک توضیح تکمیلی اضافه کرده است. حتی اگر این توضیح از نظر منطقی صحیح باشد، چون عینا یا به‌ صورت استنتاج نزدیک در contextها نیامده، RAGAS آن را ادعای بدون پشتوانه فرض کرده و امتیاز Faithfulness را کاهش داده است. این دقیقا یکی از رایج‌ ترین علت‌ های افت faithfulness در سیستم‌ های RAG حقوقی است: افزودن توضیح عمومی به‌جای بسنده کردن به متن قانون. راهکار این است که در پرامپت تولید پاسخ، صریحا الزام شود هر بند پاسخ یا (الف) نقل مستقیم از context باشد یا (ب) با برچسب استنباط و با اشاره‌ ی دقیق به ماده ارائه شود.

پس از آن، Faithfulness پایین در چند نمونه‌ ی دیگر دیده می ‌شود: از جمله سوال محدودیت‌ های صادر کننده چک برگشتی با Faithfulness = 0.771 و ابطال سند رسمی با Faithfulness = 0.797 و آثار چک برگشتی با Faithfulness = 0.849. الگوی مشترک این نمونه‌ ها معمولا یکی از این دو حالت است: یا پاسخ، فهرست‌ وار چند اثر/محدودیت را بیان کرده که بعضی از آن‌ها به‌ طور صریح در context بازیابی‌شده نیستند ، یا context های بازیابی‌ شده پوشش کامل مواد مرتبط را ندارند و مدل برای کامل کردن پاسخ، از دانش عمومی استفاده کرده است. به بیان دقیق‌ تر، این خطا ها غالبا خطای تولید نیستند؛ بلکه از یک عدم هم‌ ترازی بین دامنه سوال و پوشش context می‌ آیند: وقتی سوال دامنه‌ ی وسیع دارد، اگر retrieval فقط یک یا دو ماده را برگرداند، پاسخ کاملِ چندبخشی به ناچار شامل گزاره‌هایی می‌شود که در context نیستند و RAGAS آن را عدم وفاداری می‌ سنجد.

بنابراین علت‌ های اصلی افت امتیاز ها در نتایج را می‌ توان به‌ صورت دقیق چنین گفت که: (1) پاسخ‌ های محافظه‌ کارانه از نوع “اطلاعات کافی نیست” که از نگاه RAGAS به‌ عنوان پاسخ نامرتبط امتیاز relevancy را شدیدا کاهش می‌ دهد؛ (2) افزودن توضیح تکمیلی یا تعمیم در پاسخ‌ های تعریف‌ محور یا فهرستی که باعث افت faithfulness می‌ شود؛ (3) ناکافی بودن پوشش retrieval برای سوالات گسترده (context ناقص)، که مدل را به سمت ترکیب یا تکمیل مبتنی بر دانش عمومی هل می‌ دهد؛ و (4) در برخی سناریو ها عدم خلوص context ها  که هم faithfulness را پایین می‌ آورد و هم ممکن است تمرکز پاسخ را از سؤال دور کند.

در نهایت، بر مبنای همین خطا یابی، پیشنهاد های اصلاحی که مستقیما به بهبود همین دو معیار منجر می ‌شوند عبارت‌ اند از: افزایش کیفیت و پوشش context ها برای پرسش‌ های دامنه‌ وسیع (افزایش k یا retrieval هدفمند ماده‌ محور)، اجبار مدل به ارجاع ماده/تبصره برای هر گزاره، محدود کردن تولید به آنچه در context آمده (و جدا کردن استنباط از نقل)، و باز طراحی قالب پاسخ‌ های اطلاعات کافی نیست به‌صورت پاسخ‌ های نیمه‌ کامل و دقیقا سؤال‌ محور تا Answer Relevancy به صفر سقوط نکند. این تغییرات، با حفظ محافظه‌ کاری حقوقی سیستم، هم Faithfulness را پایدار تر می‌ کند و هم نوسان شدید Answer Relevancy را کاهش می‌ دهد.
</div>

```python
import chainlit
```

```python
app_code = r'''
# =========================
# app.py - Chainlit UI for your RAG Agent (LangGraph + LanceDB + FastEmbed)
# Based on NLP_HW5_Q1.ipynb
# Run:
#   chainlit run app.py -w --port 8000
# =========================

import os
import time
import traceback
from typing import TypedDict, List, Dict, Any, Optional, Callable

import chainlit as cl

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langgraph.graph import StateGraph, END

import lancedb
from fastembed import TextEmbedding
from pydantic import BaseModel, Field
from typing import Literal

# =========================
# Config (override with env if you want)
# =========================
# Put your lancedb_storage folder next to app.py OR set env LANCEDB_DIR to your absolute path.
LANCEDB_DIR = os.getenv("LANCEDB_DIR", "./lancedb_storage")
TABLE_NAME  = os.getenv("TABLE_NAME", "law_chunks")

# Embedding model (same as your notebook)
EMB_MODEL_NAME = os.getenv("EMB_MODEL_NAME", "intfloat/multilingual-e5-large")

# LLM model: IMPORTANT - must be supported by AvalAI
# You can override with: set OPENAI_MODEL in env
OPENAI_MODEL = os.getenv("OPENAI_MODEL", "gpt-4o-mini")

# =========================
# Agent State (same as your notebook)
# =========================
class AgentState(TypedDict, total=False):
    user_query: str
    rewritten_query: str

    intent: str
    intent_response: str

    metadata_filter: Dict[str, Any]

    retrieved_docs: List[Dict[str, Any]]
    reranked_docs: List[Dict[str, Any]]

    retry_count: int
    need_retry: bool

    final_answer: str

    timings: List[Dict[str, Any]]
    question_id: str
    source_law: str


# =========================
# Utility: timed wrapper (same logic as notebook)
# =========================
def timed(node_name: str, fn: Callable[[AgentState], AgentState]) -> Callable[[AgentState], AgentState]:
    def _wrapped(state: AgentState) -> AgentState:
        if "timings" not in state:
            state["timings"] = []

        start = time.perf_counter()
        out = fn(state)
        end = time.perf_counter()

        state["timings"].append({
            "node": node_name,
            "start": start,
            "end": end,
            "duration_ms": (end - start) * 1000.0
        })

        return out
    return _wrapped


# =========================
# LLM (AvalAI / OpenAI-compatible)
# =========================
def get_llm():
    api_key = os.getenv("OPENAI_API_KEY")
    base_url = os.getenv("OPENAI_BASE_URL")

    if not api_key:
        raise ValueError("OPENAI_API_KEY is not set")
    if not base_url:
        raise ValueError("OPENAI_BASE_URL is not set (should end with /v1)")

    return ChatOpenAI(
        model=OPENAI_MODEL,
        temperature=0,
        api_key=api_key,
        base_url=base_url,
    )


# =========================
# Embedding (FastEmbed) - same as notebook
# =========================
embedder = TextEmbedding(model_name=EMB_MODEL_NAME)

def embed_query(text: str) -> List[float]:
    return list(embedder.embed([f"query: {text}"]))[0].tolist()

def embed_passage(text: str) -> List[float]:
    return list(embedder.embed([f"passage: {text}"]))[0].tolist()


# =========================
# LanceDB connect
# =========================
db = lancedb.connect(LANCEDB_DIR)
table = db.open_table(TABLE_NAME)


# =========================
# Node 1: Rewrite Query (same intent as notebook)
# =========================
rewrite_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "تو فقط وظیفه بازنویسی پرسش کاربر را داری.\n"
        "سؤال را شفاف، دقیق و مناسب بازیابی اسناد قانونی کن.\n"
        "از خودت پاسخ نده، توضیح نده، فقط متن سؤال بازنویسی‌شده را برگردان."
    ),
    ("human", "{question}")
])

def rewrite_query_node(state: AgentState) -> AgentState:
    user_q = state["user_query"]
    llm = get_llm()
    rewritten = llm.invoke(
        rewrite_prompt.format_messages(question=user_q)
    ).content.strip()
    return {"rewritten_query": rewritten}


# =========================
# Node 2: Intent Classification (Greeting / Abusive / Law)
# =========================
intent_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "نوع پیام کاربر را دقیقاً در یکی از این سه دسته تشخیص بده.\n"
        "فقط یکی از این سه کلمه را خروجی بده و هیچ توضیح اضافه نده:\n"
        "Greeting\nAbusive\nLaw"
    ),
    ("human", "{question}")
])

def classify_intent_node(state: AgentState) -> AgentState:
    user_q = state["user_query"]
    llm = get_llm()

    label = llm.invoke(
        intent_prompt.format_messages(question=user_q)
    ).content.strip()

    if label == "Greeting":
        return {
            "intent": "greeting",
            "intent_response": "سلام 👋 خوش اومدی! سؤال حقوقی‌ات رو بپرس تا بررسی کنم."
        }

    if label == "Abusive":
        return {
            "intent": "abusive",
            "intent_response": (
                "من برای کمک طراحی شدم و وارد گفت‌وگوی توهین‌آمیز نمی‌شوم. "
                "اگر سوالت را محترمانه مطرح کنی، با کمال میل راهنمایی می‌کنم."
            )
        }

    return {"intent": "law"}


# =========================
# Node 3: Extract Metadata (structured output)
# Based on your fixed allowed titles/domains
# =========================
LAW_TITLES = Literal[
    "قانون تأمين اجتماعي",
    "قانون ثبت اسناد و املاك",
    "قانون روابط موجر و مستأجر",
    "قانون صدور چك",
    "قانون كار",
    "قانون ماليات بر ارزش افزوده",
    "قانون مبارزه با قاچاق كالا و ارز",
]

LAW_DOMAINS = Literal[
    "anti_smuggling",
    "check",
    "labor",
    "landlord_tenant",
    "registration_realestate",
    "social_security",
    "vat_tax",
]

class MetadataFilter(BaseModel):
    law_title: Optional[LAW_TITLES] = Field(default=None, description="نام قانون فقط از بین مقادیر مجاز")
    law_domain: Optional[LAW_DOMAINS] = Field(default=None, description="دامنه فقط از بین مقادیر مجاز")

    # Optional: if your DB has these fields, we can use them; otherwise they will be ignored safely
    article_no: Optional[str] = Field(default=None, description="شماره ماده (اگر قابل استخراج بود)")
    has_tabserah: Optional[bool] = Field(default=None, description="آیا تبصره ذکر شده است یا خیر")

metadata_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "از روی پرسش بازنویسی‌شده، اگر ممکن بود فیلترهای متادیتایی استخراج کن.\n"
        "قانون (law_title) فقط باید یکی از این ۷ مقدار باشد:\n"
        "- قانون تأمين اجتماعي\n- قانون ثبت اسناد و املاك\n- قانون روابط موجر و مستأجر\n- قانون صدور چك\n- قانون كار\n- قانون ماليات بر ارزش افزوده\n- قانون مبارزه با قاچاق كالا و ارز\n"
        "law_domain هم فقط یکی از این‌ها باشد:\n"
        "anti_smuggling, check, labor, landlord_tenant, registration_realestate, social_security, vat_tax\n"
        "اگر مطمئن نیستی، مقدار را null بگذار."
    ),
    ("human", "{question}")
])

def extract_metadata_node(state: AgentState) -> AgentState:
    rewritten_q = state["rewritten_query"]
    llm = get_llm()
    extractor = llm.with_structured_output(MetadataFilter)
    metadata: MetadataFilter = extractor.invoke(
        metadata_prompt.format_messages(question=rewritten_q)
    )
    return {"metadata_filter": metadata.model_dump(exclude_none=True)}


# =========================
# Node 4: Context Retrieve (LanceDB vector search + where)
# Same logic style as notebook (embedding then table.search(embedding))
# =========================
def build_where_clause(metadata_filter: dict) -> str:
    clauses = []

    # Apply only if those columns exist in your DB schema
    schema_fields = set(table.schema.names)

    if "law_title" in metadata_filter and "law_title" in schema_fields:
        clauses.append(f"law_title = '{metadata_filter['law_title']}'")

    if "law_domain" in metadata_filter and "law_domain" in schema_fields:
        clauses.append(f"law_domain = '{metadata_filter['law_domain']}'")

    if "article_no" in metadata_filter and "article_no" in schema_fields:
        clauses.append(f"article_no = '{metadata_filter['article_no']}'")

    if "has_tabserah" in metadata_filter and "has_tabserah" in schema_fields:
        val = "true" if metadata_filter["has_tabserah"] else "false"
        clauses.append(f"has_tabserah = {val}")

    return " AND ".join(clauses)

def context_retrieve_node(state: AgentState, K: int = 10) -> AgentState:
    query = state["rewritten_query"]
    metadata_filter = state.get("metadata_filter", {}) or {}

    query_embedding = embed_query(query)

    where_clause = build_where_clause(metadata_filter)

    search = table.search(query_embedding)
    if where_clause:
        search = search.where(where_clause)

    results = search.limit(K).to_list()
    return {"retrieved_docs": results}


# =========================
# Node 5: Rerank (same as notebook; score/_distance)
# =========================
def rerank_node(state: AgentState, N: int = 3, min_score: float = 0.15) -> AgentState:
    docs = state.get("retrieved_docs", [])
    retry_count = state.get("retry_count", 0)

    def get_score(doc):
        if "score" in doc:
            return float(doc["score"])
        if "_distance" in doc:
            return 1.0 / (1.0 + float(doc["_distance"]))
        return 0.0

    docs_sorted = sorted(docs, key=get_score, reverse=True)
    top_docs = docs_sorted[:N]

    best_score = get_score(top_docs[0]) if top_docs else 0.0
    relevance_ok = best_score >= min_score

    need_retry = (not relevance_ok) and (retry_count == 0)

    return {
        "reranked_docs": top_docs,
        "need_retry": need_retry
    }


# =========================
# Node 6: Generate Answer (same spirit as notebook)
# =========================
answer_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "تو یک دستیار حقوقی هستی که فقط و فقط بر اساس متن قوانین ارائه‌شده پاسخ می‌دهد.\n"
        "اگر اطلاعات کافی در متن‌های داده‌شده وجود نداشت، صریحاً بگو «اطلاعات کافی در اسناد بازیابی‌شده وجود ندارد».\n"
        "از خودت قانون، ماده یا تبصره اختراع نکن.\n"
        "در صورت امکان، در پاسخ نام قانون و شماره ماده را ذکر کن."
    ),
    (
        "human",
        "سؤال:\n{question}\n\n"
        "متن‌های قانونی بازیابی‌شده:\n{context}\n\n"
        "پاسخ نهایی:"
    )
])

def format_context_for_answer(docs: List[Dict[str, Any]]) -> str:
    if not docs:
        return ""

    parts = []
    for i, d in enumerate(docs, 1):
        law_title = d.get("law_title", "")
        article_no = d.get("article_no", "")
        text = d.get("text", "")

        header = f"[{i}]"
        if law_title:
            header += f" {law_title}"
        if article_no:
            header += f" | ماده/بخش: {article_no}"

        parts.append(f"{header}\n{text}".strip())

    return "\n\n---\n\n".join(parts)

def generate_answer_node(state: AgentState) -> AgentState:
    question = state["rewritten_query"]
    docs = state.get("reranked_docs", [])

    context_text = format_context_for_answer(docs)

    llm = get_llm()
    answer = llm.invoke(
        answer_prompt.format_messages(
            question=question,
            context=context_text
        )
    ).content.strip()

    return {"final_answer": answer}


# =========================
# Routers + helpers (same logic as notebook)
# =========================
def end_intent_node(state: AgentState) -> AgentState:
    return {"final_answer": state.get("intent_response", "")}

def bump_retry_node(state: AgentState) -> AgentState:
    return {
        "retry_count": state.get("retry_count", 0) + 1,
        "need_retry": False
    }

def intent_router(state: AgentState) -> str:
    intent = state.get("intent", "law")
    if intent in ("greeting", "abusive"):
        return "end_intent"
    return "extract_metadata"

def retry_router(state: AgentState) -> str:
    return "retry" if state.get("need_retry") else "generate"


# =========================
# Build TIMED graph (so you can still log timings if needed)
# =========================
rewrite_t = timed("rewrite_query", rewrite_query_node)
classify_t = timed("classify_intent", classify_intent_node)
extract_t  = timed("extract_metadata", extract_metadata_node)
retrieve_t = timed("context_retrieve", lambda s: context_retrieve_node(s, K=10))
rerank_t   = timed("rerank", lambda s: rerank_node(s, N=3, min_score=0.15))
generate_t = timed("generate_answer", generate_answer_node)
end_intent_t = timed("end_intent", end_intent_node)
bump_retry_t = timed("bump_retry", bump_retry_node)

graph = StateGraph(AgentState)

graph.add_node("rewrite_query", rewrite_t)
graph.add_node("classify_intent", classify_t)
graph.add_node("end_intent", end_intent_t)

graph.add_node("extract_metadata", extract_t)
graph.add_node("context_retrieve", retrieve_t)
graph.add_node("rerank", rerank_t)
graph.add_node("bump_retry", bump_retry_t)
graph.add_node("generate_answer", generate_t)

graph.set_entry_point("rewrite_query")
graph.add_edge("rewrite_query", "classify_intent")

graph.add_conditional_edges(
    "classify_intent",
    intent_router,
    {
        "end_intent": "end_intent",
        "extract_metadata": "extract_metadata",
    }
)

graph.add_edge("end_intent", END)

graph.add_edge("extract_metadata", "context_retrieve")
graph.add_edge("context_retrieve", "rerank")

graph.add_conditional_edges(
    "rerank",
    retry_router,
    {
        "retry": "bump_retry",
        "generate": "generate_answer",
    }
)

graph.add_edge("bump_retry", "context_retrieve")
graph.add_edge("generate_answer", END)

rag_app = graph.compile()


# =========================
# Chainlit UI
# =========================
@cl.on_chat_start
async def on_chat_start():
    await cl.Message(
        content=(
            "🤖 سامانه پرسش‌وپاسخ قوانین (RAG)\n\n"
            "سؤال حقوقی‌ات رو بپرس.\n"
            "اگر سؤال سلام/احوال‌پرسی باشد پاسخ کوتاه می‌گیرى؛ اگر توهین باشد پاسخ محترمانه می‌گیرى."
        )
    ).send()


@cl.on_message
async def on_message(message: cl.Message):
    question = (message.content or "").strip()
    if not question:
        await cl.Message(content="لطفاً یک سؤال وارد کنید.").send()
        return

    # Run graph safely (in thread) to avoid blocking the event loop
    try:
        state: AgentState = {
            "user_query": question,
            "retry_count": 0,
            "timings": []
        }

        result = await cl.make_async(rag_app.invoke)(state)

        answer = result.get("final_answer", "اطلاعات کافی در اسناد بازیابی‌شده وجود ندارد.")
        await cl.Message(content=answer).send()

        # Optional: store timings in the conversation for debugging (not shown by default)
        cl.user_session.set("last_timings", result.get("timings", []))

    except Exception:
        err = traceback.format_exc()
        # show a short message to user; full trace stays in terminal logs
        await cl.Message(
            content="خطایی رخ داد. لطفاً دوباره تلاش کنید. (جزئیات در ترمینال ثبت شد.)"
        ).send()
        print(err)

'''

with open("app.py", "w", encoding="utf-8") as f:
    f.write(app_code)

print("app.py file created.")
print("chainlit run app.py -w --port 8000")
```
